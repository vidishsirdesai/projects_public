# Linear Regression
### Loss Function
$L(w, w_0) = \min_{(w, w_0)} \frac{1}{n} \sum_{i = 1}^{n} (y_i - (\bar{w}^\top \bar{x} + w_0))^2$.

Where,
- $y_i = y_{actual}$.
- $\bar{w}^\top \bar{x} + w_0 = y_{\text{predicted}}$.

### Algorithm
1. Initialize `w` and `w_0`.
2. Find the gradient of `w` and `w_0` with respect to the loss function,
    - The gradient of `w` is given by, `w_grad = -((y - y_pred).dot(2 * X))/ N`.
    - The gradient of `w_0` is given by, `w_0_grad = -(2 * np.sum(y - y_pred))/ N`.
3. Compute the gradient of `w` and `w_0` into the update equation of `w` and `w_0`,
    - The update equation of `w` is, `w = w - ss * (w_grad)`.
    - The update equation of `w_0` is, `w_0 = w_0 - ss * (w_0_grad)`.
4. Repeat the steps 2 and 3 to update `w` and `w_0` for the specified number of iterations.

### Assumptions Of Linear Regression
1. Linearity of variables: The label should have linear relationship with features. 
    - Pearson Correlation Coefficient Test can be used to confirm this.
    - A Linear Regression Model can be built to see how well it performs.
    - Plotting a graph between each of the feature and the label can also help in confirming this.
2. No multi-collinearity: The features should be independent of each other.
    - A heatmap of all the features can be plotted to confirm this.
3. Normality of Errors: Errors should be normally distributed.
    - A plot of Errors can confirm this.
    - A Shapiro Test can also be performed to confirm this.
4. No homoskedasticity: The variance of Errors should be constant. In case homoskedasticity does exist, either remove the outliers, or non-linear transformation such as Box-Cox Transformation can be performed, or a Polynomial Regression Model can be built.
    - A scatter plot of Errors can be used to confirm this.
    - Goldfeld-Quandt Test can also be performed to confirm this.
5. No auto-correlation: The residuals should not be correlated to each other. This check is made when there is a Time parameter present in the dataset. In fact, Time Series Models are built instead of Linear Regression Model in case there is a Time component.
    -  A scatter plot of Errors vs. Time can be used to confirm this.
 
### Variance Inflation Factor (VIF)
VIF is used to identify the scenarios of multi-collinearity. The feature, or features with highest VIF is dropped to improve the model's performance. VIF is mathematically expressed as,
- $\text{VIF} = \frac{1}{(1 - \text{R2})}$.
- Where,
    - $\text{R2}$ = R2 Score.

The above equation outputs a value between $(0, \infty)$. The output is interpreted as,
- $\text{VIF} > 10$ is considered to be highly multi-collinear.
- $5 <= \text{VIF} <= 10$ is considered multi-collinear, but further testing should be performed by removing the feature, and see if the model's performance increases or decreases.
- $\text{VIF} = 1$ indicates that there is no multi-collinearity between the feature, and other features.

The following is how VIF is calculated,
1. The actual label is removed from the dataset.
2. A Linear Regression Model is built by considering one of the features as label.
3. The VIF Score is calculated for that feature.
4. Steps 2 and 3 are repeated for all the features present in the dataset.
5. After calculating the VIF Scores for all the features, the features with VIF greater than 5 are dropped.
6. Finally, the remaining features are used to build the Linear Regression Model.

### Polynomial Regression
Polynomial Regression is used to model non-linear relationships between features and label. This is achieved by creating new features based on existing features using polynomial terms. For example, if there is a feature $x$, Polynomial Regression might create additional features like, $x^2$, $x^3$, etc, to capture the curvature in the relationship. This creation of new features is called as Feature Explosion.

When to stop adding features, and how to choose the best model?
- Generalization and Occam's Razor: The model that generalizes the best on the test dataset is chosen, and if there are multiple such models that generalize well on the test dataset, then according to Occam's Razor, the simplest solution is chosen.
- Bias and Variance trade-off: Bias (not $w_0$) is a measure of how close the predicted value is to the actual value. Variance (not standard deviation) is how far apart are the predicted values from each other. A model with high bias will likely be a simple model. A model with high variance will likely be a complex model.
- Degree of Polynomial Equation: Degree or power of 3 or 4 is usually the sweet spot where there is a balance of, bias and variance.

### Regularization
Regularization, like VIF is another method to remove features. The Regularization term in Linear Regression is a hyperparameter, denoted by $\lambda$.
- A higher $\lambda$ value increases the penalty, leading to a simpler model.
- A lower $\lambda$ value decreases the penalty, allowing for more complex models.

The Loss Function of Linear Regression, after the regularization parameter is introduced, is mathematically expressed as,
- $\text{Loss} = \text{MSE} + \lambda * \text{Regularization}$.

There are 3 ways in which the regularization parameter can be added to the loss function,
1. Lasso Regression:
    - Lasso stands for Least Absolute Shrinkage and Selection Operator (LASSO). The Lasso Regression uses the L1 Norm penalty. This has a sparser effect, meaning it can shrink the feature weights all the way to zero, effectively removing those features from the model.
    - Mathematically, the loss function is expressed as,
        - $Loss = (\frac{1}{n}\sum_{i = 1}^{n}(y_i - \bar{y})^2) + \lambda * (\sum_{j = 1}^{d} |W_j|)$.
    - Lasso Regression is a good choice when feature selection and interpretability are desired. It can be more susceptible to overfitting. Therefore, careful hyperparameter tuning (especially the L1 penalty strength) is crucial.
2. Ridge Regression:
    - Ridge Regression is a type of regularization technique commonly used in Linear Regression to address the problem of multi-collinearity, and improve model generalizability.
    - The Ridge Regression uses the L2 norm penalty. This penalty term is proportional to the sum of the squared coefficients (weights) of the features in the model. This shrinks all feature weights but doesn't necessarily set any to zero.
    - The penalty term discourages the model from assigning very large weights to any particular feature.
    - Mathematically, the loss function is expressed as,
        - $Loss = (\frac{1}{n}\sum_{i = 1}^{n}(y_i - \bar{y})^2) + \lambda * (\sum_{j = 1}^{d}W_j^2)$.
    - Ridge Regression is a good choice when there is a need to reduce the influence of all features but don't necessarily need feature selection. It's generally less prone to overfitting compared to LASSO.
3. ElasticNet Regression:
    - ElasticNet regression is a combination of Lasso and Ridge regression. The loss function is mathematically expressed as,
        - $Loss = (\frac{1}{n}\sum_{i = 1}^{n}(y_i - \bar{y})^2) + \lambda * ((\sum_{j = 1}^{d}W_j^2) + (\sum_{j = 1}^{d} |W_j|))$.

### Hyperparameters
- Regularization parameters:
    - Regularization techniques like Lasso Regression, and Ridge Regression introduce hyperparameters that control the strength of the penalty term used to prevent overfitting. 
        - $\alpha$ in Lasso Regression controls L1 penalty strength.
        - $\alpha$ in Ridge Regression controls L2 penalty strength.
- Number of features: While not technically a hyperparameter, it is set before training, selecting the most relevant features can be crucial for avoiding overfitting, and improving model performance. Feature selection techniques can help with this. The best way to deal with this is by deploying techniques like VIF.
- Polynomial degree: The degree of the polynomial used in Polynomial Regression can be considered a model selection parameter. A higher degree can lead to overfitting, so choosing the optimal degree is important.

### Handling Underfitting And Overfitting
- Underfitting (High Bias): 
    - The following practices can be adopted to mitigate underfitting:
        - Include relevant features that might capture more complex relationships.
        - Add more features using Feature Engineering.
        - Introduce higher degree polynomial terms to fit non-linear relationships
        - Decrease the strength of Regularization Parameter ($\lambda$), or try a different Regularization Technique.
        - Gather more data.
        - Remove noise, outliers, and other inconsistencies.
- Overfitting (High Variance): 
    - The following practices can be adopted to mitigate overfitting:
        - Remove irrelevant features.
        - Increase the strength of Regularization Parameter ($\lambda$).
        - Decrease the degree of the polynomial term.
        - Implement Cross-Validation.
        - Implement Early Stopping or reduce the number of iterations.
        - Remove noise, outliers, and other inconsistencies.

# Logistic Regression
### Loss Function
$\large{\text{Log Loss} = (-y_i \times \log(y_{pred})) - (1 - y_i) \times \log(1- y_{pred})}$.

### Algorithm
1. Initialize,
    - `weights` = The weights vector.
    - `error_list` = An empty list to store the loss values at each epoch.
    - `learning_rate`.
    - `epochs` = Number of training iteration.
2. Calculate the predicted probability (`p_i`) for each data point using the Sigmoid Function.
    - `p_i = 1.0/ (1.0 + np.exp(-(np.dot(x, weight))))`.
3. Compute the Negative Log-Loss (Binary Cross-Entropy) using the predicted probability and the true label `y`,
    - `loss = -1 * (np.mean((y * np.log(p_i) + (1 - y) * np.log(1 - p_i))))`.
4. Append the calculated `loss` to `error_list`.
5. Compute the gradient of the Loss Function with respect to the `weights`,
    - `grad = (np.dot(x.T, (p_i - y)))/ x.shape[0]`.
6. Update the `weights` using the gradient descent update rule,
    - `weights = weights - (learning_rate * grad)`.
7. Repeat the steps from 2 to 6 for the specified number of training iterations (epochs).
8. Finally, return the `weights` and the `error_list`.

### Multi-Class Classification Using Logistic Regression
Say that there are 3 classes of fishes, Tuna, Salmon, and Pomfret. A Logistic Regression Model is built for each of the classes. Meaning, 3 models are built in total, Model 1 to find if a fish is Tuna or not, Model 2 to find if a fish is Salmon or not, and Model 3 to find if a fish is Pomfret or not.

Depending upon what the Model outputs the highest probability score, the fish is classified into that class. 

This technique is called as One-Versus-Rest (OvR).

### Regularization
The Regularization term in Logistic Regression is a hyperparameter, denoted by $c$.
- A higher $c$ value increases the penalty, leading to a simpler model.

Here, $c = \frac{1}{\lambda}$.

The Loss Function for Logistic Regression, after the Regularization parameter is introduced, is mathematically expressed as,
- $\text{Loss} = \text{Negative Log-Loss (NLL)} + c * \text{Regularization}$.

There are 3 methods in which the Regularization parameter can be added to the Loss Function,

This has a sparser effect, meaning it can shrink some feature weights all the way to zero, effectively removing those features from the model.
    - Mathematically, the Loss Function is expressed as,

    - Lasso Regression in a good choice when feature selection and interpretability are desired. It can be more susceptible to overfitting. Therefore, careful hyperparameter tuning (especially the L1 penalty strength) is crucial.

    - Ridge Regression is a good choice when there is a need to reduce the influence of all features but don't necessarily need feature selection. It is generally less prone to overfitting compared to LASSO.
3. ElasticNet Regression:
    - ElasticNet Regression is a combination of Lasso and Ridge Regression. The loss function is mathematically expressed as,
        - $\text{Loss} = (\text{NLL}) + c * ((\sum_{j = 1}^{d}W_j^2) + (\sum_{j = 1}^{d} |W_j|))$.

# KNN
### Algorithm
1. For a data point $x_i$, the distance between it, and every training data point in the dataset is calculated using a distance formula (Euclidean, Manhattan, Minkowski, etc). The choice of the distance metric depends on the number of dimensions.
2. After calculating the distances for all the training points, they are sorted in ascending order based on the distance from the new data point $x_i$. This ensures that the data points closest to $x_i$ are at the beginning of the sorted list.
3. The value of `k`, a hyperparameter, determines the number of nearest neighbors to consider for prediction. The `k` hyperparameter is crucial, and can be tuned based on the specific problem, and dataset. After sorting, the top `k` data points are chosen as the nearest neighbors for the new data point.
4. Prediction:
    - For classification: The most frequent class label among the `k` nearest neighbors is assigned to the new data point (majority vote).
    - For regression: The average value of the `k` nearest neighbors is used to predict the value for the new data point.

NOTE: The value of `k` chosen should be such that it should not be a multiple of the number of classes. This is done to prevent a scenario where the unknown data point is at equidistant from 2 sets of data points. But in case where, after hyperparameter tuning, the best value is a number that is a multiple of the number of classes, then in that case, either pick at random between the classes, or sum up the distances of all the points present in a particular class, do the same for all the classes, and then pick the class which has the lowest sum of distance.

### Missing Value Imputation Using KNN
KNN imputation is a popular technique for handling missing values in a dataset. It leverages the similarity between data points to estimate the missing value for a specific sample.

Steps involved,
1. Locate all the data points with missing values in the dataset.
2. Choose the number of `k` nearest neighbors to consider for imputation. A common starting point is `k = 5`.
3. Select a distance metric to measure the similarity between data points. Euclidean Distance is a common choice, but others like Manhattan Distance, or Minkowski Distance might be suitable depending on the data.
4. For each data point with a missing value, identify the `k` nearest neighbors in the dataset based on the chosen distance metric. Only consider data points with complete values for the feature with the missing value.
5. Impute the missing value. There are 2 common approaches,
    - Mean imputation: Calculate the average value of the chosen feature for the `k` nearest neighbors. This imputed value replaces the missing value in the original data point.
    - Weighted mean imputation: Assign weights to the nearest neighbors based on their distance to the point with the missing value. Closer neighbors receive higher weights, influencing the imputation more. Then, calculate the weighted average of the feature values from the neighbors to estimate the missing value.

### Hyperparameters
- Number of Neighbors (`k`): This parameter determines how many of the closest data points (neighbors) are used to make a prediction. A common strategy is to experiment with different `k` values, and evaluate the model's performance using metrics like Accuracy, Precision, Recall, or F1 Score (for classification) or MSE or R2 Score (for regression).
    - Low `k`: Can lead to overfitting, especially with noisy data, as the model might be too sensitive to the specific characteristics of the neighbors.
    - High `k`: Can lead to underfitting, as the model might not capture the local patterns well, and become overly influenced by irrelevant data points further away.
- Distance Metric: KNN relies on a distance metric to measure similarity between data points. The choice of distance metric can impact the model's performance, especially if the data has features with different scales, or distributions. Experimenting with different distance metrics might be necessary to find the best fit for the dataset. The most common choice is Euclidean Distance, but other options include,
    - Manhattan Distance: This calculates the sum of the absolute differences between corresponding features of two data points.
    - Minkowski Distance: This generalizes Euclidean and Manhattan distances, allowing for control over the emphasis placed on different feature dimensions.
- Feature Scaling: KNN relies on calculating distances between data points. If features have significantly different scales, or variances, features with larger scales might dominate the distance calculations, leading to suboptimal performance. Feature scaling techniques like Standardization (Z-Score Normalization), or Normalization (Min-Max Scaling) can help ensure all features contribute equally to the distance calculations.
- Outlier Handling: Outliers can significantly affect KNN models, especially when using a small number of neighbors (`k`). Outliers might be wrongly classified, or even pull the model's predictions in the wrong direction. Consider outlier detection and removal techniques or robust distance metrics that are less sensitive to outliers if they are a significant concern in the data.

# Decision Trees
### Algorithm
1. Start with the entire dataset at the Root Node.
2. Find the combination of feature and data point (sample) that best splits the data points into distinct groups based on the label (Classfication), or prediction of continuous value (Regression). Metrics like Entropy, or Gini Impurtiy (GI) are used to evaluate the goodness of the split.
3. Create branches for each possible outcome (outcome is the number of classes for classification, and the values for regression) of the test at the Internal Node.
4. Recursively repeat steps 2, and 3, for each branch until a stopping criteria is met. This criteria could be, reaching a certain level of purity (meaning all data points in a Leaf Node belong to the same class in Classification, or have similar target values in Regression), or reaching the maximum depth defined for the tree, or having no more informative features to split on.

### Visualizing Decision Trees
- The first node is called as the Root Node.
- The nodes that are not split further are called as Leaf Nodes.
- The nodes in between the Root Node, and the Leaf Nodes are called as Internal Nodes, or Decision Nodes.
- The decision boundaries in a Decision Tree are axis parallel. Meaning, a node that splits the data linearly, polynomially, or logistically, cannot be created.
- Pure Nodes, or Homogenous Nodes occur when there is only 1 class present as a result of a split.

### How To Decide What Feature To Split On, And What Threshold To Split On?
- Say that there are 200 data points, 100 positive, and 100 negative. Also, say that there are $d$ features.
- Now say that, 2 features $j$ and $k$ are selected out of the $d$ features. Also say that 2 thresholds $\tau_j$ and $\tau_k$ are chosen.
- Consider case 1, where the feature $j$ and the threshold $\tau_j$ is splitting the 200 data points into, 10 positives and 90 negatives if the decision was yes, and 90 positives and 10 negatives if the decision was no.
- Consider case 2, where the feature $k$ and the threshold $\tau_k$ is splitting the 200 data points into, 60 positives and 40 negatives if the decision was yes, and 40 positives and 60 negatives if the decision was no.
- The better combination of feature, and threshold that should be selected is case 1. Why? Because, this choice is resulting in a purer node, or better segregation.

### How Many Decisions Should Be Compared?
Given,
- $d$ = Number of features.
- $n$ = Number of unique values per feature.

For each feature, $(n - 1)$ threshold values can be potentially considered. For instance, if a feature has 1, 2, and 3, thresholds between 1 and 2 and between 2 and 3 can be considered. Therefore, the total number of possible decisions to evaluate for a single feature is $(n - 1)$.

$\text{Total number of decisions compared} = d \times (n - 1)$.

### Entropy, Gini Impurity And Information Gain
- Entropy:
    - Entropy essentially measures the randomness, or uncertainty associated with classifying a data point at a node.
    - A higher entropy (closer to 1) indicates a more impure node. The probabilities of different classes are closer, making it harder to predict the class label for a new data point.
    - A lower entropy (closer to 0) indicates a purer node. The data points in the node predominantly belong to one class, leading to higher confidence in predicting the class label.
    - It reaches its maximum value of 1 when the class distribution is 50/50.
- Gini Impurity:
    - Gini Impurity measures how often a randomly chosen element from the set would be incorrectly classified if it was randomly labeled according to the distribution of labels in the subset.
    - A value of 0 indicates perfect purity.
    - It reaches its maximum value of 0.5 when the class distribution is 50/50.
- Information Gain:
    - The drop from impurity at root level, and the impurtiy after the split is called as Information Gain. Low Entropy or Gini Impurity, and high Information Gain (IG) are desired.

### Pruning
Pruning is cutting unnecessary branches in Decision Trees. It is done to prevent overfitting on train set. The hyperparameters associated with Pruning are,
1. `max_depth`: Defines the depth of the tree.
2. `min_sample_split`: Defines the number of minimum data points required to split. For example, if it is set to 600, any node that contains data points lower than this will not be split.
3. `min_sample_leaf`: Defines the minimum number of data points every node should have after splitting.
4. `max_leaf_nodes`: Defines the maximum number of leaf nodes that a node can be split into.

### Hyperparameters
- Splitting criteria:
    - This parameter determines how the decision tree chooses the best feature and split point at each node for creating the tree structure. Common options include,
        - Gini Impurity (for classification): Measures the impurity, or heterogeneity within a node. A lower Gini value indicates a more homogenous node (better split).
        - Entropy (for classification): Another measure of node impurity based on information theory.
        - Mean Squared Error (for regression): Measures the average squared difference between actual target values and the predicted values. Lower MSE indicates a better split for regression tasks.
        The choice of the criteria depends on the kind of task (classification v. regression) and the preference for handling impurity.
- Maximum depth (`max_depth`):
    - This parameter controls the maximum allowable depth of the tree. Deeper trees can capture more complex relationships but are also more prone to overfitting. Here's how depth affects the tree,
        - Shallow tree: Less complex, may underfit, and miss important patterns in the data.
        - Deep tree: More complex, can capture intricate details but might overfit the training data, and performs poorly on unseen data.
    - Finding the right balance between capturing complexity, and avoiding overfitting is crucial. Techniques like cross-validation can be used to evaluate performance with different maximum depth values.
- Minimum samples per split (`min_sample_split`):
    - This parameter sets the minimum number of data points required to split a node. Lower values can lead to overfitting as the tree might capture irrelevant details in smaller splits. Higher values can lead to underfitting if the tree becomes too shallow.
- Minimum samples per split (`min_sample_leaf`):
    - This parameter sets the minimum number of data points allowed in a final leaf node (terminal node). Lower values can lead to overfitting as the model might create overly specific leaf nodes with very few data points. Higher values can result in underfitting as the tree might merge similar data points into single leaves, potentially missing important distinctions.
- Maximum features per split (`max_features`):
    - This parameter determines how many features are considered at each node for splitting. Considering all features at each split can lead to complex trees and overfitting. Considering a random subset of features at each split can help reduce overfitting and improve generalization.
- Tuning techniques:
    - There is no single best set of hyperparameters for a decision tree. The optimal configuration depends on the dataset and the specific task. The following are some techniques for tuning hyperparameters,
        1. `GridSearchCV`: Evaluates a predefined grid of hyperparameter values and selects the combination leading to best performance.
        2. `RandomizedSearchCV`: Samples hyperparameter values randomly from a defined range and selects the best performing combination. This can be more efficient than grid search for large datasets.
        3. Cross-Validation: Splits the data into training and validation sets. Trains the model with different hyperparameter combination on the training data and evaluates performance on the validation set.

### Handling Underfitting And Overfitting
- Underfitting (High Bias)
    - The following practices can be adopted to mitigate underfitting:
        - Increase tree complexity:
            - Allowing the tree to grow deeper or increasing the number of branches can help it capture more complex patterns
            - This involves adjusting hyperparameters like `max_depth`.
        - Reduce constraints:
            - Lowering the values of hyperparameters like `min_samples_leaf` and `min_samples_split` can allow the tree to create more complex rules.
        - Add more features using Feature Engineering.
- Overfitting (High Variance): 
    - The following practices can be adopted to mitigate overfitting:
        - Pruning:
            - This involves removing branches of the tree that do not significantly contribute to its predictive power.
            - Techniques:
                - Cost-complexity pruning: Removes branches based on a complexity parameter.
                - Reduced error pruning: Removes branches if they decrease accuracy on a validation set.
        - Limit tree depth:
            - Setting a maximum depth for the tree prevents it from growing too complex.
            - This is controlled by the `max_depth` hyperparameter.
        - Minimum samples per leaf:
            - Requiring a minimum number of samples in each leaf node prevents the tree from creating overly specific rules.   
            - This is controlled by the `min_samples_leaf` hyperparameter.
        - Minimum samples split:
            - Requiring a minimum number of samples to split an internal node.   
            - This is controlled by the `min_samples_split` hyperparameter.
        - Remove irrelevant features.
        - Implement Cross-Validation.
        - Remove noise, outliers, and other inconsistencies.

### Time Complexity And Space Complexity
- Time complexity:
    - Training: $O(n \times m \times \log(n))$.
    - Prediction: $O(d \times \log(n))$.
- Space complexity:
    - Training: $O(d \times m)$.
    - Prediction: $O(1)$.
- Where,
    - $n$ = Number of data points in the training set.
    - $m$ = Number of features in the data.
    - $d$ = Depth of the Decision Tree.

# Random Forest
### Algorithm
1. Create a Bootstrap dataset which is of the same size as the original dataset:
    - Consider that the following dataset is used to build a Random Forest,

        | # | chest_pain | good_blood_circulation | blocked_arteries | weight | heart_disease |
        | :-: | :-: | :-: | :-: | :-: | :-: |
        | 1 | No | No | No | 125 | No |
        | 2 | Yes | Yes | Yes | 180 | Yes |
        | 3 | Yes | Yes | No | 210 | No |
        | 4 | Yes | No | Yes | 167 | Yes |
    
    - To create a Bootstrapped dataset that is of the same size as the original, samples are selected at random, with replacement (a sample if selected once, can be selected again). The Bootstrapped dataset looks as follows,

        | # | chest_pain | good_blood_circulation | blocked_arteries | weight | heart_disease |
        | :-: | :-: | :-: | :-: | :-: | :-: |
        | 2 | Yes | Yes | Yes | 180 | Yes |
        | 1 | No | No | No | 125 | No |
        | 4 | Yes | No | Yes | 167 | Yes |
        | 4 | Yes | No | Yes | 167 | Yes |
    
    - Observe that the 4th sample has been selected twice.
2. Create a Decision Tree using the Bootstrapped dataset, but only use a random subset of features (or columns) at each step:
    - In this example, 2 variables, viz., `good_blood_circulation` and `blocked_arteries` are considered at each step (selected randomly).
    - For the sake of this example, `good_blood_circulation` was found to do the best job of splitting the samples at the Root Node.

        ```mermaid
        flowchart TD
            A[good_blood_circulation]
            B[1]
            C[2]
            A --> B
            A --> C
        ```
    
    - At 1, another set of 2 random variables are considered to perform the split. Consider that, `chest_pain` was found to be the best. Therefore,

        ```mermaid
        flowchart TD
            A[good_blood_circulation]
            B[chest_pain]
            C[1]
            D[2]
            E[2]
            F[1]
            G[2]
            A --> B
            A --> E
            B --> C
            B --> D
            E --> F
            E --> G
        ```
    - This process of splitting is carried out until the Leaf Nodes are reached.

To sum up, a Decision Tree is created using,
1. Bootstrapped dataset, and
2. only a random subset of variables were considered at each step.

The above 2 steps are repeated to build multiple Decision Trees. This results in a wide variety of Trees. The variety is what makes Random Forests more effective than individual Decision Trees.

### Performing The Task Of Classification
Consider the following new sample,

| chest_pain | good_blood_circulation | blocked_arteries | weight | heart_disease |
| :-: | :-: | :-: | :-: | :-: |
| Yes | No | No | 168 |  |

To know if the above patient has a heart disease or not, the above data is run down the first Decision Tree created in the Random Forest. Consider that, this sample reaches the leaf node which says "Yes".

Similarly, the sample is run down all the Decision Trees, and the value in the leaf node is recorded.

In the end, the class with the majority vote is assigned as the class to which the sample belongs to.

### Bagging
Bootstrapping the samples in the dataset plus using the aggregate to make a decision is called as "Bagging".

### Evaluation Of Random Forest Using Out-Of-Bag Score
Since dupicates are allowed in the Bootstrapped dataset, a few of the entries end up not being included in the Bootstrapped dataset. Typically, about 1/3rd of the original data does not end up in the Bootstrapped dataset.

The following is the entry that did not end up in the Bootstrapped dataset in the example used,

| # | chest_pain | good_blood_circulation | blocked_arteries | weight | heart_disease |
| :-: | :-: | :-: | :-: | :-: | :-: |
| 3 | Yes | Yes | No | 210 | No |

NOTE: If the original dataset were larger, there would be more than 1 entry.

This is called as the Out-Of-Bag sample, and the dataset is called Out-Of-Bag dataset. The following is how Out-Of-Bag dataset is used to evaluate the Random Forest,
- Since the Out-Of-Bag sample was not used to create the tree. This can be run through the first tree to check if the Decision Tree correctly classifies the sample or not.
- This is repeated for all the other trees that were built without it.
- This process is repeated for all the other Out-Of-Bag samples for all the trees in the Random Forest.
- Ultimately, the Accuracy of Random Forest can be measured by measuring the proportion of Out-Of-Bag samples that were correctly classified by the Random Forest. The proportion of Out-Of-Bag samples that were incorrectly classified is the Out-Of-Bag Error.
- Finally, the Out-Of-Bag Error can be used to compare different Random Forests built using different number of random subsets of variables used to split the node. For example, Random Forest built using 2 variables per step vs. Random Forest built using 3 variables per step, etc. A bunch of different settings can be tested to choose the most accurate Random Forest.

### Hyperparameters
- `n_estimators`: The number of trees in the Forest. More Trees generally improve performance but increase training time.
-  `max_samples`: Row samples ratio (0 to 1). Defines the proportion of rows (data points) randomly chosen for training each Tree.
- `max_features`: Column sampling (values: `"log2"`, `"sqrt"`, `None`), Defines the number of features randomly chosen at each split in a Tree.
- `ccp_alpha` (CCP Alpha (Cost Complexity Pruning)): This hyperparameter penalizes the model based on the number of terminal nodes (leaves). Fewer nodes generally lead to simpler, less overfitting models.
    - Higher `ccp_alpha` values lead to more pruning (fewer nodes), potentially causing underfitting.
    - Lower values allow for more complex Trees (more nodes), potentially leading to overfitting.

# AdaBoost
### Introduction
- In Random Forest, each time a Tree is built, it is a full sized Tree. Some Trees might be bigger than the others, but there is no pre-determined maximum depth.
- In contrast, in a Forest of Trees built with AdaBoost, the Trees are usually just a Node, and 2 Leaves.
- A Tree with just 1 Node and 2 Leaves is called a Stump. So AdaBoost is a Forest of Stumps rather than Trees. Stumps are not great at making accurate classifications. For example, if the following dataset were used to determine if someone has a heart disease or not,

    | # | chest_pain | good_blood_circulation | blocked_arteries | weight | heart_disease |
    | :-: | :-: | :-: | :-: | :-: | :-: |
    | 1 | No | No | No | 125 | No |
    | 2 | Yes | Yes | Yes | 180 | Yes |
    | 3 | Yes | Yes | No | 210 | No |
    | 4 | Yes | No | Yes | 167 | Yes |

    then a full sized Decision Tree would take advantage of all 4 variables, to make a decision. But, a Stump can use only use 1 variable to make a decision. Thus, Stumps are called "weak learners".
- In a Random Forest, each Tree has an equal vote on the final classification. In contrast, in a Forest of Stumps built with AdaBoost, some Stumps get more say in the final classification than others.
- Lastly, in a Random Forest, each Decision Tree is built independently of the others. Meaning, it does not matter which Tree is built first. In contrast, in a Forest of Stumps built with AdaBoost, order is important. The errors that the first Stump makes, influence how the second Stump is built. Similarly, the errors that the second Stump makes, influence how the third Stump is built, and so on.

### Algorithm
1. Create a Forest of Stumps:
    - Consider the following data,

        | # | chest_pain | blocked_arteries | patient_weight | heart_disease |
        | :-: | :-: | :-: | :-: | :-: |
        | 1 | Yes | Yes | 205 | Yes |
        | 2 | No | Yes | 180 | Yes |
        | 3 | Yes | No | 210 | Yes |
        | 4 | Yes | Yes | 167 | Yes |
        | 5 | No | Yes | 156 | No |
        | 6 | No | Yes | 125 | No |
        | 7 | Yes | No | 168 | No |
        | 8 | Yes | Yes | 172 | No |
    
    - The objective is to create a Forest of Stumps with AdaBoost to predict if a patient has a heart disease. The prediction will be made using patient's `chest_pain`, `blocked_arteries`, and `patient_weight`.
    - The first thing that is done is, a weight (`sample_weight`) is assigned to each sample which indicates how important it is to be correctly classified. At start, all the samples get the same weight, which is given by, $\frac{1}{\text{Total Number of Samples}}$. In this case, it is $\frac{1}{8}$, which makes all the samples equally important. However, after the first Stump is built, these weights will change in order to guide how the next Stump is created. The dataset looks as follows,

        | # | chest_pain | blocked_arteries | patient_weight | heart_disease | sample_weight |
        | :-: | :-: | :-: | :-: | :-: | :-: |
        | 1 | Yes | Yes | 205 | Yes | 1/8 |
        | 2 | No | Yes | 180 | Yes | 1/8 |
        | 3 | Yes | No | 210 | Yes | 1/8 |
        | 4 | Yes | Yes | 167 | Yes | 1/8 |
        | 5 | No | Yes | 156 | No | 1/8 |
        | 6 | No | Yes | 125 | No | 1/8 |
        | 7 | Yes | No | 168 | No | 1/8 |
        | 8 | Yes | Yes | 172 | No | 1/8 |
    
    - The first Stump in the Forest is built by finding the variable that does the best job at classifying the samples. Because all the weights are the same, `sample_weights` can be ignored for now.
    - The best variable will the one whose Gini Index will be the least. Consider that, the best variable that does the best job at classifying was found to be `patient_weight`. Therefore, this will be the first Stump in the Forest.
2. Determine the amount of say that the Stump will have in the final classification:
    - This is done based on how well the Stump classified the samples. Consider that the first Stump made one error were the following sample was incorrectly classified,

        | # | chest_pain | blocked_arteries | patient_weight | heart_disease | sample_weight |
        | :-: | :-: | :-: | :-: | :-: | :-: |
        | 4 | Yes | Yes | 167 | Yes | 1/8 |
    
    - Meaning, the Stump classified this patient to not have a heart disease, where as the patient has a heart disease.
    - The Total Error for a Stump is the sum of the weights associated with incorrectly classified samples. In this case, the Total Error = $\frac{1}{8}$.

        NOTE: Because all of the `sample_weight` add up to 1, Total Error will always be between 0, for a perfect Stump, and 1, for a horrible Stump.

    - Total Error is used to determine the Amount of Say a Stump has in the final classification with the following formula,
        - $\text{Amount of Say} = \frac{1}{2} \log(\frac{1 - \text{Total Error}}{\text{Total Error}})$.
    - A graph of the Amount of Say can be plotted by plugging in a bunch of numbers between 0 and 1 for Total Error, like so,
    
        <img src = "../07_supervised_learning/ensemble_learning/amount_of_say.png" width = "400" height = "300">

    - The blue line tells the Amount of Say for Total Error values between 0 and 1. 
        - When a Stump does a good job, and the Total Error is small, then the Amount of Say is relatively large, positive value.
        - When a Stump is no better at classification than flipping a fair coin, and Total Error = 0.5, then the Amount of Say will be 0.
        - When a Stump does a bad job, and the Total Error is close to 1, then the Amount of Say is relatively large, negative value.
    - Therefore for the first stump, $\text{Amount of Say} = \frac{1}{2} \log(\frac{1 - \frac{1}{8}}{\frac{1}{8}}) = 0.97$.
3. Modify the `sample_weights`:
    - Now to modify the weights so that the next Stump will into account take Errors that the current Stump made, the `sample_weight` of the incorrectly classified sample should be increased, and all other `sample_weights` of all other samples should be decreased.
    - To increase the `sample_weight` of the incorrectly classified sample, the following formula is used,
        - $\text{New Sample Weight} = \text{Sample Weight} \times e^{\text{Amount of Say}}$.
        - Therefore,
            - $\text{New Sample Weight} = \frac{1}{8} \times e^{0.97} = 0.33$.
    - To decrease the `sample_weight` of the correctly classified sample, the following formula is used,
        - $\text{New Sample Weight} = \text{Sample Weight} \times e^{-\text{Amount of Say}}$.
        - Therefore,
            - $\text{New Sample Weight} = \frac{1}{8} \times e^{-0.97} = 0.05$.
    - The new dataset looks as follows,

        | # | chest_pain | blocked_arteries | patient_weight | heart_disease | sample_weight | new_sample_weight |
        | :-: | :-: | :-: | :-: | :-: | :-: | :-: |
        | 1 | Yes | Yes | 205 | Yes | 1/8 | 0.05 |
        | 2 | No | Yes | 180 | Yes | 1/8 | 0.05 |
        | 3 | Yes | No | 210 | Yes | 1/8 | 0.05 |
        | 4 | Yes | Yes | 167 | Yes | 1/8 | 0.33 |
        | 5 | No | Yes | 156 | No | 1/8 | 0.05 |
        | 6 | No | Yes | 125 | No | 1/8 | 0.05 |
        | 7 | Yes | No | 168 | No | 1/8 | 0.05 |
        | 8 | Yes | Yes | 172 | No | 1/8 | 0.05 |
    
    - The `new_sample_weight` column needs to be normalized so that it will add up to 1. This is done by finding the column sum of `new_sample_weight` column and then dividing each value in the `new_sample_weight` column by that sum. The resulting dataset will be,

        | # | chest_pain | blocked_arteries | patient_weight | heart_disease | sample_weight | new_sample_weight |
        | :-: | :-: | :-: | :-: | :-: | :-: | :-: |
        | 1 | Yes | Yes | 205 | Yes | 1/8 | 0.07 |
        | 2 | No | Yes | 180 | Yes | 1/8 | 0.07 |
        | 3 | Yes | No | 210 | Yes | 1/8 | 0.07 |
        | 4 | Yes | Yes | 167 | Yes | 1/8 | 0.49 |
        | 5 | No | Yes | 156 | No | 1/8 | 0.07 |
        | 6 | No | Yes | 125 | No | 1/8 | 0.07 |
        | 7 | Yes | No | 168 | No | 1/8 | 0.07 |
        | 8 | Yes | Yes | 172 | No | 1/8 | 0.07 |

    - The dataset after considering the `new_sample_weight` will look as follows,

        | # | chest_pain | blocked_arteries | patient_weight | heart_disease | sample_weight |
        | :-: | :-: | :-: | :-: | :-: | :-: |
        | 1 | Yes | Yes | 205 | Yes | 0.07 |
        | 2 | No | Yes | 180 | Yes | 0.07 |
        | 3 | Yes | No | 210 | Yes | 0.07 |
        | 4 | Yes | Yes | 167 | Yes | 0.49 |
        | 5 | No | Yes | 156 | No | 0.07 |
        | 6 | No | Yes | 125 | No | 0.07 |
        | 7 | Yes | No | 168 | No | 0.07 |
        | 8 | Yes | Yes | 172 | No | 0.07 |
4. Create a new Bootstrapped sample dataset for the next Stump,
    - In theory, the Sample Weights could be used to calculate Weighted Gini Indexes to determine which variable should split the next Stump.
    - Alternatively, a new collection of samples can be created that contains duplicate copies of the samples with the largest Sample Weights. This is done by 
        - creating a new, but empty dataset that is of the same size as the original.
        - then a random number is chosen between 0 and 1.
        - then depending on where that number falls, the sample_weight column is used like a distribution. 
            - if the number is between 0 and 0.07, then,
        
                | # | chest_pain | blocked_arteries | patient_weight | heart_disease | sample_weight |
                | :-: | :-: | :-: | :-: | :-: | :-: |
                | 1 | Yes | Yes | 205 | Yes | 0.07 |

                is selected into the new collection of samples.
            - if the number is between 0.07 and 0.14, then,
        
                | # | chest_pain | blocked_arteries | patient_weight | heart_disease | sample_weight |
                | :-: | :-: | :-: | :-: | :-: | :-: |
                | 2 | No | Yes | 180 | Yes | 0.07 |

                is selected into the new collection of samples.
            - and so on.
    - Consider that the new sample collection looks as follows after doing the above,

        | # | chest_pain | blocked_arteries | patient_weight | heart_disease | sample_weight |
        | :-: | :-: | :-: | :-: | :-: | :-: |
        | 5 | No | Yes | 156 | No | 0.07 |
        | 4 | Yes | Yes | 167 | Yes | 0.49 |
        | 6 | No | Yes | 125 | No | 0.07 |
        | 4 | Yes | Yes | 167 | Yes | 0.49 |
        | 4 | Yes | Yes | 167 | Yes | 0.49 |
        | 8 | Yes | Yes | 172 | No | 0.07 |
        | 1 | Yes | Yes | 205 | Yes | 0.07 |
        | 4 | Yes | Yes | 167 | Yes | 0.49 |
    
    - Ultimately, the sample,

        | # | chest_pain | blocked_arteries | patient_weight | heart_disease | sample_weight |
        | :-: | :-: | :-: | :-: | :-: | :-: |
        | 4 | Yes | Yes | 167 | Yes | 0.49 |
    
        is added 4 times to the collection, which reflects its `sample_weights`.
    - After this, the orignal sample are gotten rid of. Meaning, the following are the samples which will be used to build the new Stump (observe that equal sample weights are assigned to the new samples),

        | # | chest_pain | blocked_arteries | patient_weight | heart_disease | sample_weight |
        | :-: | :-: | :-: | :-: | :-: | :-: |
        | 1 | No | Yes | 156 | No | 1/8 |
        | 2 | Yes | Yes | 167 | Yes | 1/8 |
        | 3 | No | Yes | 125 | No | 1/8 |
        | 4 | Yes | Yes | 167 | Yes | 1/8 |
        | 5 | Yes | Yes | 167 | Yes | 1/8 |
        | 6 | Yes | Yes | 172 | No | 1/8 |
        | 7 | Yes | Yes | 205 | Yes | 1/8 |
        | 8 | Yes | Yes | 167 | Yes | 1/8 |
    
    - The next Stump will be built using the above samples.
5. Repeat:
    - This whole process is repeated from the top.
    - This is how, the Errors that the first Tree makes, influences how the second Tree is built. Similarly, the Errors that the second Stump makes, influence how the third Stump is built, and so on.

### Classification Using The Forest Of Stumps Created Using AdaBoost
Consider that,
- a group of Stumps have classified a patient to have heart disease, and the Total Amount of Say for that group is 2.7.
- another group of Stumps have classified a patient to not have heart disease, and the Total Amount of Say for that group is 1.23.

The patient is classified as "has heart disease", because the first group's Amount of Say total is greater than the second group's Amount of Say total.

# Gradient Boosting Decision Trees (GBDT)
### Introduction
- GBDT is very similar to AdaBoost. Consider,

    | height (m) | favorite_color | gender | weight (kg) |
    | :-: | :-: | :-: | :-: |
    | 1.6 | Blue | Male | 88 |
    | 1.6 | Green | Female | 76 |

- If the above dataset were to be used to predict weight, then AdaBoost starts by building a very short Tree, called a Stump, from the training data, and the amount of say that the Stump has on the final output is based on how well it compensated for the previous errors. Then AdaBoost builds the next Stump based on errors that the previous Stump made. AdaBoost continues to build Stumps in this fashion until it has build the number of Stumps that were specified, or it has a perfect fit.
- In contrast, GBDT starts by building a single Leaf, instead of a Tree or Stump. This Leaf represents an initial guess for the weights of all the samples. When trying to predict a continuous value like weight, the first guess is the average value. Then GBDT builds a Tree. Like AdaBoost, this Tree is based on the errors made by the previous Tree, but unlike AdaBoost, this Tree is usually larger than a Stump. That said, GBDT still restricts the size of the Tree.
- In the example following, Trees with upto 4 leaves are built, but no larger. However, in practice, the maximum number of leaves are set between 8 and 32. 
- Thus, like AdaBoost, GBDT builds fixed sized Trees based on the previous Tree's errors, but unlike AdaBoost, each Tree can be larger than a Stump. Also like AdaBoost, GBDT scales the Trees. However, GBDT scales all Trees by the same amount. Then GBDT builds another Tree based on the errors made by the previous Tree, and then it scales the Tree, and GBDT continues to build Trees in this fashion until it has made the number of trees that were specified, or until the Trees fail to improve the fit.

### Algorithm For Regression
1. Calculate the average of the label (i.e., weight).
    - $\text{Average} = 71.2$.
    - This is the first attempt at predicting everyone's weight. In other words, if the process were stopped right now, the prediction of everyone's weight would be 71.2 kg. However, GBDT does not stop here.
2. Build a Decision Tree based on the Errors from the first Tree:
    - The next thing to do is to build a Tree based on the errors from the first Tree. The errors that the previous tree made are the differences between the observed weights, and the predicted weights, 71.2. Therefore, $\text{Error} = \text{Observed Value} - \text{Predicted Value}$. The difference is termed as Pseudo Residual, and is saved in a new column.

        NOTE: The term Pseudo Residual is based on Linear Regression, where the difference between the observed values, and the predicted values results in Residuals.
    
    - The dataset after computing the Pseudo Residuals, would look as follows,

        | height (m) | favorite_color | gender | weight (kg) | residual |
        | :-: | :-: | :-: | :-: | :-: |
        | 1.6 | Blue | Male | 88 | 16.8 |
        | 1.6 | Green | Female | 76 | 4.8 |
        | 1.5 | Blue | Female | 56 | -15.2 |
        | 1.8 | Red | Male | 73 | 1.8 |
        | 1.5 | Green | Male | 77 | 5.8 |
        | 1.4 | Blue | Female | 57 | -14.2 |

    - Now a Tree will be built using `height`, `favorite_color`, and `gender` to predict the Residuals.
    - Say that the Tree looks as follows,

        ```mermaid
        flowchart TD
            A[gender = F]
            B[height < 1.6]
            C[-14.2, -15.2]
            D[4.8]
            E[color != Blue]
            F[1.8, 5.8]
            G[16.8]
            A --> B
            A --> E
            B --> C
            B --> D
            E --> F
            E --> G
        ```
    
    - By restricting the total number of Leaves, fewer Leaves than Residuals are obtained. As a result, the follwing 2 rows of data go to the same Leaf,

        | height (m) | favorite_color | gender | weight (kg) | residual |
        | :-: | :-: | :-: | :-: | :-: |
        | 1.5 | Blue | Female | 56 | -15.2 |
        | 1.4 | Blue | Female | 57 | -14.2 |
    
    - Therefore, the Residuals are replaced by average residuals, $\frac{-15.2 - 14.2}{2} = -14.7$.
    - Also, the following 2 rows of data go to the same Leaf,

        | height (m) | favorite_color | gender | weight (kg) | residual |
        | :-: | :-: | :-: | :-: | :-: |
        | 1.8 | Red | Male | 73 | 1.8 |
        | 1.5 | Green | Male | 77 | 5.8 |

    - Therefore, the Residuals are replaced by average residuals here also, $\frac{1.8 + 5.8}{2} = 3.8$.
    - The Tree now looks as follows,

        ```mermaid
        flowchart TD
            A[gender = F]
            B[height < 1.6]
            C[-14.7]
            D[4.8]
            E[color != Blue]
            F[3.8]
            G[16.8]
            A --> B
            A --> E
            B --> C
            B --> D
            E --> F
            E --> G
        ```
    - The original Leaf can now be combined with the new Tree to make a new prediction of an individual's weight from the training data. The following is how it is done,
        - Start with the initial prediction, $71.2$, 
        - Then the data at the end of the Tree in the Leaf of the new Tree is multiplied with the learning rate, and is added with the initial prediction, $71.2 + (0.1 \times 16.8) = 72.9$. The prediction has now improved slightly compared to the initial prediction.
        - This is repeated for all the values in the Leaf nodes.
    - Scaling the Tree by learning rate results in a small step in the right direction.
    - According to the bloke that invented Gradient Boost, Jerome Freidman, empirical evidence shows that taking a lot of small steps in the right direction results in better predictions with a testing dataset, i.e., lower Variance.
    - The dataset at the after building the first DT looks as follows,

        | height (m) | favorite_color | gender | weight (kg) | residual |
        | :-: | :-: | :-: | :-: | :-: |
        | 1.6 | Blue | Male | 88 | (88 - (71.2 + (0.1 * 16.8))) = 15.1 |
        | 1.6 | Green | Female | 76 | (76 - (71.2 + (0.1 * 4.8))) = 4.3|
        | 1.5 | Blue | Female | 56 | 56 - (71.2 + (0.1 * -15.2)) = -13.7 |
        | 1.8 | Red | Male | 73 | 73 - (71.2 + (0.1 * 1.8)) = 1.4 |
        | 1.5 | Green | Male | 77 | 77 - (71.2 + (0.1 * 5.8)) = 5.4 |
        | 1.4 | Blue | Female | 57 | 57 - (71.2 + (0.1 * -14.2)) = -12.7 |

    - Observe that the new `residuals` (pseudo residuals) are all smaller than the original `residual`.
    - The dataset can be beautified as,

        | height (m) | favorite_color | gender | weight (kg) | residual |
        | :-: | :-: | :-: | :-: | :-: |
        | 1.6 | Blue | Male | 88 | 15.1 |
        | 1.6 | Green | Female | 76 | 4.3|
        | 1.5 | Blue | Female | 56 | -13.7 |
        | 1.8 | Red | Male | 73 | 1.4 |
        | 1.5 | Green | Male | 77 | 5.4 |
        | 1.4 | Blue | Female | 57 | -12.7 |
3. Build the next Tree to predict the new Residuals:
    - Say that the Tree is as follows,

        ```mermaid
        flowchart TD
            A[gender = F]
            B[height < 1.6]
            C[-12.7, -13.7]
            D[4.3]
            E[color != Blue]
            F[1.4, 5.4]
            G[15.1]
            A --> B
            A --> E
            B --> C
            B --> D
            E --> F
            E --> G
        ```

        NOTE: In this simple example the branches are the same as before. However, in practice, the trees can be different each time.
    
    - In the above Tree, since multiple samples ended up in the same Leaves, the average of them is calculated. Therefore, the Tree will now look like,

        ```mermaid
        flowchart TD
            A[gender = F]
            B[height < 1.6]
            C[-13.2]
            D[4.8]
            E[color != Blue]
            F[3.4]
            G[16.8]
            A --> B
            A --> E
            B --> C
            B --> D
            E --> F
            E --> G
        ```

    - This new Tree is combined with the previous Tree, and the initial Leaf to calculate the residuals.

        NOTE: All the Trees are scaled by learning rate, which in this case is 0.1.
    
    - The new prediction is made as, $71.2 + (0.1 \times 16.8) + (0.1 \times 15.1) = 74.4$.
    - New Residuals are calculate as, the initial Leaf plus the scaled values from the first Tree plus the scaled values from the second Tree. Therefore,
        - $88 - (71.2 + (0.1 \times 16.8) + (0.1 \times 15.1)) = 13.6$.
    - The dataset after the second Tree is built will look like,

        | height (m) | favorite_color | gender | weight (kg) | residual |
        | :-: | :-: | :-: | :-: | :-: |
        | 1.6 | Blue | Male | 88 | 13.6 |
        | 1.6 | Green | Female | 76 | 3.9|
        | 1.5 | Blue | Female | 56 | -12.4 |
        | 1.8 | Red | Male | 73 | 1.1 |
        | 1.5 | Green | Male | 77 | 5.1 |
        | 1.4 | Blue | Female | 57 | -11.4 |
4. Repeat the above steps for the specified number of Trees:
    - Each time a Tree is added to the prediction, the residuals get smaller. This process of building DTs on residuals, and then computing the new residuals is repeated until the specified number of Trees have been built, or until the Trees fail to improve the fit.
5. Prediction:
    - For a new data point (sample), the weight can be predicted by, starting at the initial prediction, then adding the scaled value from the first Tree, and the second Tree, and the third Tree, and so on.
    - Once the computation is all done, the predicted weight is obtained.

### Algorithm For Classification
1. Compute the initial prediction:
    - At first, similar to GBDT for regression, a Leaf that represents an initial prediction for every individual is built. When Gradient Boost is used for classification, the initial prediction for every individual is the $\log(\text{odds})$. Think of $\log(\text{odds})$ as the Logistic Regression equivalent of the average.
    - Consider the following dataset,

        | likes_popcorns | age | favorite_color | loves_trolls_2 |
        | :-: | :-: | :-: | :-: |
        | Yes | 12 | Blue | Yes |
        | Yes | 87 | Green | Yes |
        | No | 44 | Blue | No |
        | Yes | 19 | Red | No |
        | No | 32 | Green | Yes |
        | No | 14 | Blue | Yes |

    - To calculate the overall $\log(\text{odds})$,
        - 4 people love Trolls 2, and 2 people do not, $\log(\text{odds})$ is, $\log(\frac{4}{2}) = 0.7$ (Natural Logarithm).
    - The above will be the initial prediction.
    - How is the above used for classification?
        - Just like with Logistic Regression, the easiest way to use the $\log(\text{odds})$ for classification is to convert it to a probability. This is done using Logistic Function.
        - Therefore, $\text{P}(\text{Loving Trolls 2}) = \frac{e^{\log(\text{odds})}}{1 + e^{\log(\text{odds})}} = 0.7$.
        - Since the probability of loving Trolls 2 is greater than 0.5, everyone in the training dataset can be classified as someone who loves Trolls 2.
    
            NOTE: While 0.5 is a very common threshold for making classification decisions based on Probability, a different value could have been used as well.

    - Now, classifying everyone in the training dataset as someone who loves Troll 2 is obviously not correct, because 2 of the people do not love the movie. Pseudo Residuals between the observed value and the predicted value can be calculated to measure the goodness of the prediction,
        - $\text{Pseudo Residual} = \text{Observed Value} - \text{Predicted Value}$.
    - The dataset looks like,

        | likes_popcorns | age | favorite_color | loves_trolls_2 | residual |
        | :-: | :-: | :-: | :-: | :-: |
        | Yes | 12 | Blue | Yes | (1 - 0.7) = 0.3 |
        | Yes | 87 | Green | Yes | (1 - 0.7) = 0.3 |
        | No | 44 | Blue | No | (0 - 0.7) = -0.7 |
        | Yes | 19 | Red | No | (0 - 0.7) = -0.7 |
        | No | 32 | Green | Yes | (1 - 0.7) = 0.3 |
        | No | 14 | Blue | Yes | (1 - 0.7) = 0.3 |

    - Therefore,

        | likes_popcorns | age | favorite_color | loves_troll_2 | residual |
        | :-: | :-: | :-: | :-: | :-: |
        | Yes | 12 | Blue | Yes | 0.3 |
        | Yes | 87 | Green | Yes | 0.3 |
        | No | 44 | Blue | No | -0.7 |
        | Yes | 19 | Red | No | -0.7 |
        | No | 32 | Green | Yes | 0.3 |
        | No | 14 | Blue | Yes | 0.3 |
    
    - Now a DT is built using `likes_popcorn`, `age`, and `favorite_color` to predict the residuals. The Tree looks as follows,

        ```mermaid
        flowchart TD
            A[color = Red]
            B[-0.7]
            C[age > 37]
            D[0.3, -0.7]
            E[0.3, 0.3, 0.3]
            A --> B
            A --> C
            C --> D
            C --> E
        ```

        NOTE: Similar to GBDT for regression, the number of Leaves in the Tree are limited to 3.
    
    - In case of GBDT for regression, a Leaf with single residual had an output value equal to that residual. In contrast, when GBDT is used for classification, the situation is a little more complex. This is because the predictions are in terms of the $\log(\text{odds})$, and the Leaf is derived from a probability. Hence, they can't be added together to get a new $\log(\text{odds})$ prediction without some sort of transformation. The formula for transformation is given by,
        - $\frac{\sum \text{Residual}_i}{\sum [\text{Previous Probability}_i \times (1 - \text{Previous Probability}_i)]}$.
    - Therefore,
        - $\frac{-0.7}{[0.7 \times (1 - 0.7)]} = -3.3$.
        - $\frac{0.3 - 0.7}{(0.7 \times (1 - 0.7)) + (0.7 \times (1 - 0.7))} = -1$.
        - $\frac{0.3 + 0.3 + 0.3}{(0.7 \times (1 - 0.7)) + (0.7 \times (1 - 0.7)) + (0.7 \times (1 - 0.7))} = 1.4$.
    - The Tree now looks like,

        ```mermaid
        flowchart TD
            A[color = Red]
            B[-3.3]
            C[age > 37]
            D[-1]
            E[1.4]
            A --> B
            A --> C
            C --> D
            C --> E
        ```
    
    - Now the predictions can be updated by combining the initial Leaf with the new Tree.

        NOTE: Just like before, the new Tree is scaled by a learning rate.
    
    - This example uses a relative large learning rate for illustrative purposes. However, 0.1 is more common.
    - Now, the $\log(\text{odds})$ for the first person in the dataset is predicted as,
        - $\log(\text{odds}) = 0.7 + (0.8 * 1.4) = 1.8$.
    - And the predicted probability is calculated using Logistic Function as,
        - $\text{Probability} = \frac{e^{1.8}}{1 + e^{1.8}} = 0.9$.
    - Therefore, a small step is taken in the right direction.
    - Similarly, the same can be repeated for all the samples. The dataset look as follows,

        | likes_popcorns | age | favorite_color | loves_troll_2 | predicted_proba |
        | :-: | :-: | :-: | :-: | :-: |
        | Yes | 12 | Blue | Yes | 0.9 |
        | Yes | 87 | Green | Yes | 0.5 |
        | No | 44 | Blue | No | 0.5 |
        | Yes | 19 | Red | No | 0.1 |
        | No | 32 | Green | Yes | 0.9 |
        | No | 14 | Blue | Yes | 0.9 |

    - The new residuals are,

        | likes_popcorns | age | favorite_color | loves_troll_2 | predicted_proba | residual |
        | :-: | :-: | :-: | :-: | :-: | :-: |
        | Yes | 12 | Blue | Yes | 0.9 | 0.1 |
        | Yes | 87 | Green | Yes | 0.5 | 0.5 |
        | No | 44 | Blue | No | 0.5 | -0.5 |
        | Yes | 19 | Red | No | 0.1 | -0.1 |
        | No | 32 | Green | Yes | 0.9 | 0.1 |
        | No | 14 | Blue | Yes | 0.9 | 0.1 |

    - The new dataset looks as follows,

        | likes_popcorns | age | favorite_color | loves_troll_2 | residual |
        | :-: | :-: | :-: | :-: | :-: |
        | Yes | 12 | Blue | Yes | 0.1 |
        | Yes | 87 | Green | Yes | 0.5 |
        | No | 44 | Blue | No | -0.5 |
        | Yes | 19 | Red | No | -0.1 |
        | No | 32 | Green | Yes | 0.1 |
        | No | 14 | Blue | Yes | 0.1 |
    
    - Now that the residuals have been calculated, a new Tree can be built.
3. Repeat:
    - This process is repeated until the specified number of Trees have been built, or until the Trees fail to improve the fit.
4. Prediction:
    - For a new data point (sample), the prediction is made by, starting at the initial prediction, then adding the scaled value from the first Tree, and the second Tree, and the third Tree, and so on.
    - The above will output the $\log(\text{odds})$, which is then converted to a probability value by applying the Logistic Function.
    - The classification that a person loves Trolls 2 or not is decided based on the set threshold. In this case, the threshold is 0.5. Therefore,
        - if probability > 0.5, then the person loves Trolls 2.
        - if probability < 0.5, then the person does not love Trolls 2.

# Naive Bayes
### Algorithm
1. Find the frequency of occurrance of all the words:
    - Imagine a scenario where the objective is to classify messages into spams, and hams.
    - To begin with, histograms of all the words occurring in ham are plotted. Consider that the following are the words with their frequency of occurrence,

        | word | frequency |
        | :-: | :-: |
        | Dear | 8 |
        | Friend | 5 |
        | Lunch | 3 |
        | Money | 1 |

2. Calculate the probabilities of seeing each word given that they belong to a particular class: 
    - This can be used to calculate the probabilities of seeing each word, given that it was a ham. For example, the probability of the word "Dear" occurring in ham is, $P(\text{Dear} | \text{ham}) = \frac{\text{Number of occurrences of the words}}{\text{Total words in ham}} = \frac{8}{17} = 0.47$. Similarly, the same can be done for the rest of the words. The dataset looks as follows after the computation,

        | word | frequency | P(word given ham) |
        | :-: | :-: | :-: |
        | Dear | 8 | 0.47 |
        | Friend | 5 | 0.29 |
        | Lunch | 3 | 0.18 |
        | Money | 1 | 0.06 |
    
    - The same is also done with the words occurring in spam. Consider that the following are the words with their frequency of occurrence,

        | word | frequency |
        | :-: | :-: |
        | Dear | 2 |
        | Friend | 1 |
        | Lunch | 0 |
        | Money | 4 |

    - The following is how the above dataset will look like after the probabilities of seeing each word, given that it was a spam,

        | word | frequency | P(word given spam) |
        | :-: | :-: | :-: |
        | Dear | 2 | 0.29 |
        | Friend | 1 | 0.14 |
        | Lunch | 0 | 0.00 |
        | Money | 4 | 0.57 |
    
    - NOTE: 
        - Total number of words in ham = 17.
        - Total number of words in spam = 7.
    - Because the probabilities of discrete, individual words has been calculated, and not the probability of something continuous, like weight, or height, these probabilities of are also called as "likelihoods".
3. Classify a new message:
    - Now consider that, a new message is received that said, "Dear Friend". To classify it as spam, or ham,
        1. Begin with an initial guess about the probability that any message, regardless of what it says, is a ham. This guess can be any probability value, but a common guess is estimated from the training data. For example, consider that 8 of the 12 messages are hams, then the initial probability guess will be 0.67. Therefore, $P(\text{ham}) = 0.67$. This initial guess is called as "prior probability".
        2. Next, the initial guess is multiplied by the probability of the word "Dear" occurring given that the message is ham, and the probability of the word "Friend" occurring given that the message is ham, $P(\text{ham}) \times P(\text{Dear} | \text{ham}) \times P(\text{Friend} | \text{ham}) = 0.67 \times 0.47 \times 0.29 = 0.09$. 0.09 is proportional to the probability that the message is a ham, given that it says "Dear Friend".
        3. Next, an initial guess is made about the probability that any message, regardless of what it says, is a spam. This guess can be any probability value, but a common guess is estimated from the training data. For example, consider that 4 of the 12 messages are spams, then the initial probability guess will be 0.33. Therefore, $P(\text{spam}) = 0.33$.
        4. Next, the initial guess is multiplied by the probability of the word "Dear" occurring given that the message is spam, and the probability of the word "Friend" occurring given that the message is spam, $P(\text{spam}) \times P(\text{Dear} | \text{spam}) \times P(\text{Friend} | \text{spam}) = 0.33 \times 0.29 \times 0.14 = 0.01$. 0.01 is proportional to the probability that the message is a spam, given that it says "Dear Friend".
        5. Since, $P(\text{ham} | \text{Dear Friend}) > P(\text{spam} | \text{Dear Friend})$, the message is classified as "ham".
    - Now consider that the message is "Lunch Money Money Money Money". Since the probability of seeing the word "Money" is much higher in spam, than in ham, it seems reasonable to predict this message will end up being spam. The calculation is as follows,
        1. $P(\text{ham}) = 0.67$.
        2. $P(\text{ham}) \times P(\text{Lunch} | \text{ham}) \times P(\text{Money} | \text{ham}) \times P(\text{Money} | \text{ham}) \times P(\text{Money} | \text{ham}) \times P(\text{Money} | \text{ham}) = 0.67 \times 0.18 \times (0.06)^4 = 0.000002$.
        3. $P(\text{spam}) = 0.67$.
        4. $P(\text{spam}) \times P(\text{Lunch} | \text{spam}) \times P(\text{Money} | \text{spam}) \times P(\text{Money} | \text{spam}) \times P(\text{Money} | \text{spam}) \times P(\text{Money} | \text{spam}) = 0.33 \times 0 \times (0.57)^4 = 0$.
    - This tells that, if a message contains the word "Lunch", it will never be classified as spam, and it will always be classified as ham. Ideally, this should not be the case. Therefore, to work around this problem, 1 additional count is added to each word in ham, and spam. The dataset after doing so looks as follows,
    - Ham data,

        | word | frequency |
        | :-: | :-: |
        | Dear | 9 |
        | Friend | 6 |
        | Lunch | 4 |
        | Money | 2 |
    
    - Spam data,

        | word | frequency |
        | :-: | :-: |
        | Dear | 3 |
        | Friend | 2 |
        | Lunch | 1 |
        | Money | 5 |
    
    - The number of counts added to each word is typically represented by $\alpha$. This method of increasing the count is called as Laplace Smoothing.
    - The new probabilities look as follows,
        - Ham data,

            | word | frequency | P(word given ham) |
            | :-: | :-: | :-: |
            | Dear | 9 | 0.43 |
            | Friend | 6 | 0.29 |
            | Lunch | 4 | 0.19 |
            | Money | 2 | 0.10 |

        - Spam data,

            | word | frequency | P(word given spam) |
            | :-: | :-: | :-: |
            | Dear | 3 | 0.27 |
            | Friend | 2 | 0.18 |
            | Lunch | 1 | 0.09 |
            | Money | 5 | 0.45 |
        
    - Recalculating the values,
        1. $P(\text{ham}) = 0.67$.
        2. $P(\text{ham}) \times P(\text{Lunch} | \text{ham}) \times P(\text{Money} | \text{ham}) \times P(\text{Money} | \text{ham}) \times P(\text{Money} | \text{ham}) \times P(\text{Money} | \text{ham}) = 0.67 \times 0.19 \times (0.10)^4 = 0.000001$.
        3. $P(\text{spam}) = 0.67$.
        4. $P(\text{spam}) \times P(\text{Lunch} | \text{spam}) \times P(\text{Money} | \text{spam}) \times P(\text{Money} | \text{spam}) \times P(\text{Money} | \text{spam}) \times P(\text{Money} | \text{spam}) = 0.33 \times 0.09 \times (0.45)^4 = 0.00122$.
        5. Since, $P(\text{ham} | \text{Dear Friend}) < P(\text{spam} | \text{Dear Friend})$, the message is classified as "spam".
- NOTE: Naive Bayes is called so, because it treats all word orders the same. Treating all word orders equally is very different to how the actual communication happens. Meaning, every language has grammar rules and common phrases, but Naive Bayes ignores all of that. Instead, Naive Bayes treats language like it is a bag full of words and each message is random handful of them. Contrary to all of this, Naive Bayes tends to perform surprisingly well while classifying spam, and ham.

### Hyperparameters
- Smoothing parameter $\alpha$:
    - Used in Laplace Smoothing to address the zero probability problem.
    - A small value (e.g., 1) is typically used to add a small amount of weight to each feature count, preventing zeros. Tuning this parameter can help balance model robustness (avoiding zero probabilities) and overfitting.
- Feature selection:
    - Naive Bayes assumes independence between features, which might not always be true.
    - Selecting a relevant subset of features that contribute most to classification can improve both bias and variance.
    - Techniques like chi-square tests, information gain, or feature importance analysis can help identify these features.
- Model selection: There are several variants of Naive Bayes, each suited for different data types,
    - Bernoulli NB: Suitable for binary features (presence or absence).
    - Multinomial NB: Suitable for features with multiple distinct values (frequency).
    - Gaussian NB: Suitable for continuous features (assumes a Gaussian distribution).
- Class priors:
    - Naive Bayes uses class priors ($P(y = c)$) to represent the probability of each class appearing in the data.
    - By default, these are estimated based on the class frequency in the training data (uniform for balanced datasets).
    - In some cases, domain knowledge about the prior probabilities of differnt classes might be available.
    - Setting informative priors based on this knowledge can potentially improve classifcation accuracy.
- Tuning techniques:
    - Techniques like `GridSearchCV` or `RandomSearchCV` can be used to explore differnt hyperparameter combinations and identify the best performing configuration for the specific dataset.
    - Cross-Validation is crucial to evaluate the model's performance on unseen data and avoid overfitting.

### Handling Underfitting And Overfitting
- Underfitting (High Bias):
    - Feature Engineering:
        - Naive Bayes relies on the independence assumption, which might not hold. Creating new features or transforming existing ones to better align with this assumption can improve performance.   
        - Ensure that features are relevant and informative for the classification task.
    - Feature Selection:
        - Removing irrelevant or redundant features can prevent the model from being diluted by noise.   
        - Use techniques like:
            - Chi-square tests.
            - Information gain.
            - Feature importance analysis.
    - Model Selection:
        - Ensure you're using the appropriate Naive Bayes variant for your data type.
            - Gaussian Naive Bayes for continuous data.   
            - Multinomial Naive Bayes for count data.   
            - Bernoulli Naive Bayes for binary data.   
        - If the selected model is not well fit for the data, the model will underfit.
    - Adjust Class Priors:
        - If you have knowledge of the class distribution that differs from the observed frequencies in the training data, adjusting the class priors can improve accuracy.
    - Gather More Data:
        - In some cases, underfitting can simply be due to insufficient training data. Collecting more data can provide the model with a better understanding of the underlying patterns.
- Overfitting (High Variance):
    - Smoothing (Laplace or Lidstone Smoothing):
        - The smoothing parameter (alpha) is crucial for preventing zero probabilities, especially with small datasets.   
        - Overfitting can occur when the model learns noise from irrelevant features.   
        - Strict feature selection can mitigate this by focusing on the most informative features.
    - Cross-Validation:
        - Use cross-validation techniques (e.g., k-fold cross-validation) to obtain a more reliable estimate of the model's performance on unseen data.   
        - This helps identify overfitting early on.
    - Simplify Features:
        - If possible, simplify or bin continuous features, or reduce the number of categories in categorical features.
    - Regularization effects of Smoothing:
        - It is worth noting, that the smoothing parameter, while addressing the issue of zero probabilities, has a regularization effect. Increasing the smoothing parameter value, effectively reduces the models variance.

- Space complexity:

    - Prediction: $O(d \times c)$.
- Where,
    - $n$ = Number of data points in the training set.
    - $d$ = Number of features in the data.
    - $c$ = Number of classes.

# Support Vector Machines
### Algorithm
1. Maximum Margin Classifier:
    - Consider a dataset which contains the information about the mass of a bunch of mice. The plot of it looks as follows,

        <img src = "../artifacts/svm1.png" width = "1080" height = "300">
    
    - The red dots represent the mice that are not obese, and the green dots represent the mice that are obese. Based on these observations, a threshold can be chosen, such that a new sample can be classified as obese if the mass of the new sample is greater than the threshold, or not obese if the mass of the new sample is lesser than the threshold.
    - But what if a new sample's mass is much closer to the threshold, but a little greater than the threshold?
        - Ideally, it should be classified as obese, because the mass is greater than the threshold.
        - But, it is actually much closer to the not obese samples.
        - Therefore, this threshold is rather too controversial, and a better mechanism for classification is required.
    - Going back to the original training dataset, the observations on the edge of the each cluster can be considered, and the midpoint between them can be used as the threshold for classification.
    - Now, when a new sample falls on the left side of the threshold, it will be closer to the observations that are not obese, than it is to the obese observations. Hence, it makes sense to classify this new observation as not obese
    - The shortest distance between the threshold and the observations is called as "margin". Since the threshold is put at halfway between the 2 observations, the distances between the observations, and the threshold are the same, and both reflect the margin. When the threshold is halfway between the 2 observations, the margin is as large as it can be. For example, if the threshold is moved to the left by a bit, then the distance between the threshold and the observation that is not obese would be smaller, and thus, the margin would be smaller than it was before. And if the threshold were to be moved to the right by a bit, then the distance between the threshold and the observation that is obese would be smaller, and thus the margin would be smaller. When the threshold that gives the largest margin to make classifications is used, it would be called as "Maximum Margin Classifier".
    - Maximum Margin Classifier is good, but what if the training data looked like the following,

        <img src = "../artifacts/svm2.png" width = "1080" height = "300">

        where there is an outlier observation that was classified as not obese, but was much closer to the obese observations.
    - In such cases, the Maximum Margin Classifier would be much closer to the obese observations, and really far from the majority of the not obese observations.
    - Now, if a new sample which is a little to left to the outlier not obese observation, would be classified as not obese, even though most of the not obese observations are much further away than the obese observations. Therefore, Maximum Margin Classifiers are super sensitive to outliers in the training data and that makes them pretty lame. Hence, a better mechanism is needed.
2. Soft Margin Classifier:
    - To have a threshold that is not so sensitive to the outliers, misclassifications must be allowed. For example, if the threshold is set such that it is halfway between like so,
    
        <img src = "../artifacts/svm2.png" width = "1080" height = "300">

        then the outlier not obese observation would be misclassified as obese. However, for a new sample which is obese, and which is a little to the left of the outlier not obese observation, is classified as obese correctly, and that makes sense, as it is closer to the obese observations.
    - Choosing a threshold that allows misclassifications is an example of the Bias-Variance tradeoff that plagues all of ML models. In other words, before misclassifications were allowed, a threshold that was very sensitive to the training data was chosen (low bias), and it performed poorly when a new sample came in (high variance). In contrast, when a threshold that was less sensitive to the training data, and allowed misclassifications was chosen (high bias), it performed better when a new sample came in (low variance).
    - When misclassifications are allowed, the distance between the observations, and the threshold is called as "Soft Margin".
    - So, how to decided if one Soft Margin is better than the other?
        - Use Cross-Validation to determine how many misclassifications, and observations to allow inside of the Soft Margin to get the best classification.
        - For example, if CV determined that a Soft Margin was the best Soft Margin, then 1 misclassification would be allowed, and 2 observations, that are correctly classified, to be within the Soft Margin.
    - When Soft Margin is used to determine the location of a threshold, then the classification model used to classify the observations is called "Soft Margin Classifier", or "Support Vector Classifier". The name "Support Vector Classifier" comes from the fact that the observations on the edge, and within the Soft Margin are called as Support Vectors.
3. Support Vector Classifier:
    - If each observation had a mass measurement, and a height measurement, the the data would be 2D. When the data are 2D, a Support Vector Classifier is a line, and in this case, the Soft Margin is measured from the following 2 points,

        <img src = "../artifacts/svm3.png" width = "400" height = "300">

    - The blue parallel lines give a sense of where all of the other points are in relation to the Soft Margin. Only 1 red sample is inside the Soft Margin and, is misclassified. Just like before, Cross-Validation is used to determine that allowing this misclassification results in better classification in the long run.
    - Now if, each observation has a mass, height and an age, then the data would be 3D.
        
        NOTE: The axis that age is on, is supposed to represent depth. The circles that are larger represent younger samples, and the circles that are smaller represent older population.

        <img src = "../artifacts/svm4.png" width = "400" height = "300">
    
    - When the data are 3D, the Support Vector Classifier forms a plane, instead of a line, and new observations are classified by determining which side of the plane they are on.

        NOTE: If mass, height, age, and blood pressure were measured, the data would be in 4D. 
    
    - Shape of SVC:
        - When the data are 1D, the Support Vector Classifier is a single point on a 1D number line. In mathematical jargon, a point is a "flat affine 0D subspace". 
        - And when the data are in 2D, the Support Vector Classifier is a 1D line in a 2D space. In mathematical jargon, line is a "flat affine 1D subspace".
        - And when the data are 3D, the Support Vector Classifier is a 2D plane in a 3D space. In mathematical jargon, plane is a "flat affine 2D subspace".
        - And when the data are in 4 or more dimensions, the Support Vector Classifier is a hyperplane. In mathematical jargon, hyperplane is a "flat affine subspace".
        - NOTE: All flat affine subspaces are called hyperplanes. So, 1D line is a hyperplane. But, the term hyperplane is only used when a shape cannot be drawn on paper.
    - Support Vector Classifiers seem pretty good because they can handle outliers, and because they allow misclassifications, they can handle overlapping classifications. But what if the training data looked as follows where there is a ton of overlap,

        <img src = "../artifacts/svm5.png" width = "1080" height = "300">
    
    - In this new example, with tons of overlap, drug dosage is measured, red dots represent patients that were not cured, and the green dots represent the patients that were cured. In other words, the drug doesn't work if the dosage is too small, or too large. Now no matter where the classifier is placed, there will be a lot of misclassifications. So Support Vector Classifiers are only semi-cool, since they don't perform well on this type of data.
    - This is were Support Vector Machines come into play.
4. Support Vector Machines:
    - At first a y-axis is added such that a graph can be drawn. The x-axis in this graph will be the dosages that have already been observed, and the y-axis will be the square of the dosages. So, for an observation with dosage = 0.5 on the x-axis, the y-axis value will be $\text{dosage}^2 = 0.5^2 = 0.25$. The same is repeated for the remaining observations. Since each observation has x and y-axis coordinates, the data are now 2D. And now that the data are 2D, a Support Vector Classifier can be drawn that separates the people who are cured from the people who are not cured, and the Support Vector Classifier can be used to classify new observations. The classification is done by squaring the dosage value, if the $\text{dosage}^2$ lies on the other side of the Support Vector Classifier, then it would be classified as not cured.

    - The main ideas behind Support Vector Machines are,
        1. Start with the data in a relatively low dimension.
        2. Move the data into a higher dimension.
        3. Find the Support Vector Classifier that separates the higher dimensional data into 2 groups.
    - How does a Kernel Function systematically finds Support Vector Classifiers in higher dimensions?
5. Kernel:
    - For this example, the Polynomial Kernel is used, which has a parameter $d$, which stands for the degree of the polynomial. 
        - When d = 1, Polynomial Kernel computes the relationships between each pair of observations in 1D, and these relationships are used to find a Support Vector Classifier.
        - When d = 2, a second dimension based on $\text{dosage}^2$ is obtained, and the Polynomial Kernel computes the 2D relationships between each pair of observations, and those relationships are used to find a Support Vector Classifier.
        - When d = 3, then a third dimension based on $\text{dosage}^3$ is obtained, and the Polynomial Kernel computes the 3D relationships between each pair of observations, and those relationships are used to find a Support Vector Classifier.
        - When d = 4 or more, the even more dimensions are obtained to find a Support Vector Classifier.
    - In summary, the Polynomial Kernel systematically increases dimensions by setting d, the degree of the polynomial, and the relationships between each pair of observations are used to find a Support Vector Classifier. Lastly, a good value for d can be found using Cross Validation.
    - Another very commonly used Kernel is the Radial Kernel, also known as the Radial Basis Function (RBF) Kernel. Unfortunately, the Radial Kernel finds Support Vector Classifiers in infinite dimensions. Therefore, an example to explain what it does would be hard to construct. However, when using it on a new observation, the Radial Kernel behaves like a Weighted Nearest Neighbor model. In other words, the closest observations (the nearest neighbors) have a lot of influence on how the new observation is classified, and observations that are further away have relatively little influence on the classification.
    - Although the examples above show the data being transformed from a relatively low dimension, to a relatively high dimension, Kernel functions only calculate the relationships between every pair of points as if they are in the higher dimensions; they don't actually do the transformation. This trick, calculating the high-dimensional relationships without actually transforming the data to the higher dimension, is called as "The Kernel Trick". The Kernel Trick reduces the amount of computation required for Support Vector Machines by avoiding the math that transforms the data from a low to a high dimensions, and it makes calculating relationships in the infinite dimensions used by Radial Kernel possible.
6. Summary:
    - However, regardless of how the relationships are calculated, the concepts are the same. When there are 2 categories, but no obvious linear classifier that separates them in a nice way, Support Vector Machines work by moving the data into a relatively high dimensional space, and finding a relatively high dimensional Support Vector Classifier that can effectively classify the observations.

### Hyperparameters
- Kernel:
    - Function: The kernel function defines how the data is transformed before applying the linear decision boundary.
    - Common choices:
        - Linear: Suitable for linearly separable data.
        - Polynomial: Can handle non-linear data by transforming it into higher dimesions.
        - RBF (Radial Basis Function): A popular choice for non-linear data, offering good flexibility.
    - Impact: Choosing the right kernel significantly affects the model's ability to learn complex relationships in the data.
- Regularization parameter ($C$):
    - Function: Controls the trade-off between maximizing the margin and allowing for misclassifications.
    - Higher $C$: Enforces a larger margin, potentially leading to stricter classification but might be less robust to noise or outliers.
    - Lower $C$: Allows for more misclassification but might result in smaller margin and potentially lower performance on unseen data.
    - Impact: Tuning $C$ is crucial to achieve a balance between accuracy and generalization.
- Gamma (for RBF kernel):
    - Function: Controls the influence of a single data point on the overall decision boundary (only relevant for the RBF kernel).
    - Higher Gamma: Leads to a more complex decision boundary, potentially fitting the training data well but at the risk of overfitting.
    - Lower Gamma: Results in a smoother decision boundary, potentially leading to better generalization but might underfit complex data.
    - Impact: Finding the right gamma value helps the model capture the appropriate level of detail in the data.
- Epsilon ($\epsilon$):
    - Function: Introducted in Soft Margin SVMs, it defines the tolerance for misclassifications.
    - Higher Epsilon: Allows for more errors, potentially useful for noisy data.
    - Lower Epsilon: Less tolerant of errors, aiming for stricter classification.
    - Impact: Epsilon helps control the flexibility of the model and the number of misclassifications allowed during training.

### Handling Underfitting And Overfitting
- Underfitting (High Bias):
    - Kernel Complexity:
        - If the data is not linearly separable, using a linear kernel will result in underfitting.
        - Consider using non-linear kernels (e.g., RBF, polynomial) to capture complex relationships.
    - Increase Model Complexity:
        - If the $C$ parameter is set too low, the model can be to simple. Increasing the $C$ parameter can allow the model to better fit the data.
    - Feature Engineering:
        - Create new features or transform existing ones to provide the model with more relevant information.
    - Gather More Data:
        - In some cases, underfitting can be due to insufficient training data.
- Overfitting (High Variance):
    - Regularization ($C$ Parameter):
        - The $C$ parameter controls the trade-off between maximizing the margin and minimizing the training error.
        - A smaller $C$ value leads to a wider margin, allowing some misclassifications, which can help prevent overfitting.   
        - A larger $C$ value attempts to classify all training examples correctly, potentially leading to overfitting.
    - Kernel Selection:
        - The choice of kernel significantly impacts the model's complexity.
        - Simpler kernels (e.g., linear) are less prone to overfitting than complex kernels (e.g., RBF).
        - If using non-linear kernels, carefully tune their parameters (e.g., 'gamma' for RBF).
    - Cross-Validation:
        - Use techniques like k-fold cross-validation to evaluate the model's performance on unseen data and identify overfitting.
        - This helps in selecting optimal hyperparameters.
    - Feature Selection/Dimensionality Reduction:
        - Reducing the number of features can simplify the model and prevent it from learning noise.
        - Techniques like PCA (Principal Component Analysis) can be used.

### Bias And Variance
- Bias
    - Definition: Bias is the error resulting from overly simplistic assumptions in the learning algorithm. A high bias model pays very little attention to the training data and oversimplifies the problem. It consistently misses the bullseye in the same direction.
    - Analogy: Imagine you're throwing darts with your eyes partially closed and using a very rigid throwing motion. You'll likely consistently miss the center in a predictable way. This consistent error is like high bias.
    - Characteristics of High Bias Models:
        - Underfitting: The model performs poorly on the training data itself and also generalizes poorly to unseen data. It fails to capture the underlying patterns in the data.
        - Simple Models: Often associated with linear models when the underlying relationship is non-linear, or decision trees with very few splits.
        - Low Sensitivity to Training Data: Changing the training data significantly won't drastically change the learned model.
    - Example: Trying to fit a straight line to a clearly curved dataset will result in high bias. The line will never accurately capture the curvature, leading to significant errors even on the data it was trained on.
- Variance
    - Definition: Variance is the error resulting from the model's sensitivity to small fluctuations in the training data. A high variance model pays too much attention to the training data, including the noise, and learns complex patterns that might not generalize well to new data. It's like your shots are scattered all over the target.
    - Analogy: Imagine you're a very excitable dart thrower. Your throws are highly influenced by minor distractions or slight variations in your stance. Your shots might sometimes hit close to the bullseye, but they'll be very inconsistent and spread out. This inconsistency is like high variance.
    - Characteristics of High Variance Models:
        - Overfitting: The model performs very well on the training data but poorly on unseen data. It has learned the noise and specific details of the training set that don't generalize.
        - Complex Models: Often associated with highly non-linear models, deep decision trees, or models with a large number of parameters.
        - High Sensitivity to Training Data: Small changes in the training data can lead to significantly different learned models.
    - Example: Training a very deep decision tree on a small dataset might lead to perfect classification of the training data, including any noisy data points. However, this complex tree is likely to perform poorly on new, unseen data because it has learned the noise instead of the true underlying patterns.
- Bias-Variance Trade-Off
    - The goal in building a good machine learning model is to find a balance between bias and variance. This is known as the bias-variance trade-off.
    - Decreasing Bias often Increases Variance: Making the model more complex (e.g., adding more features, increasing model depth) can reduce bias by allowing it to capture more intricate relationships in the training data. However, this increased complexity can also make the model more sensitive to the noise in the training data, leading to higher variance.
    - Decreasing Variance often Increases Bias: Making the model simpler (e.g., using fewer features, reducing model depth, applying regularization) can reduce variance by making it less sensitive to the specifics of the training data. However, this simplification might lead to higher bias if the model becomes too rigid to capture the underlying patterns.
- Visual Representation:
    - Imagine the bullseye is the ideal prediction.
    - High Bias, Low Variance: Shots are clustered together but far from the bullseye. The model consistently misses the true target due to oversimplification.
    - Low Bias, High Variance: Shots are scattered widely around the bullseye. On average, the shots might be centered, but individual predictions are highly variable and sensitive to the training data.
    - High Bias, High Variance: Shots are scattered far from the bullseye. The model is both making consistent errors due to oversimplification and is highly sensitive to the training data.
    - Low Bias, Low Variance: Shots are tightly clustered around the bullseye. This is the ideal scenario where the model accurately captures the underlying relationship and generalizes well.

### Regularization
Regularization is a technique used in ML to prevent overfitting by introducing a penalty term to the Loss Function. The penalty term discourages the model from learning overly complex patterns that might fit the training data too well but perform poorly on unseen data.

### Cross-Validation
- The number of data points are reduced when the dataset set is split into training, validation, and testing sets.
- In Cross-Validation (CV), the dataset is split into k parts. The choice of k is an open one.
- Consider that k is set to k = 4, then the dataset is divided into 4 equal parts. 3 parts are used for training, 1 part is used for validation. This process is repeated for 4 times, and each time a new combination of parts are used for training.
- The performance of the model is determined by the average of the performance metric used for evaluation (R2 Score, Accuracy, etc).

- This applies to any ML algorithm where distance metrics (Euclidean, Manhattan, Cosine, etc) are used. Scaling, if not done correctly, the feature that is at a higher scale will become dominant which inturn affects the predictions.

### Feature Encoding
- One of the ironclad requirements to apply ML is that the features and the labels should be of numerical data type.
- Features which are of categorical data type, if present, must be encoded.
- Feature Encoding leads to addition of new features (columns) to the dataset. Consider for example, if a dataset has the features, $f_1$, $f_2$, $f_3$, and $f_4$. Now consider thar, $f_4$ represents the blood group. If this feature is encoded using OHE, the result will add 8 more dimensions to the dataset. Making the total number of features = 11. Some of the ML algorithms become slower as the number of dimensions increase.
- Therefore, it is very important to select the correct type of Feature Encoding technique to encode the features.
- The different Feature Encoding techniques are:
    - Label Encoding: A unique integer is assigned to each category. For example, the colors, red, green, and blue may be encoded as 0, 1, and 2 respectively. But the problem with this is that, unintended ordering can be introduced if the model interprets the numbers as a ranking.
    - Ordinal Encoding: Numerical values are assigned based on the natural order of the categories, assuming one exists. T-shirt sizes "S", "M", and "L" may be encoded as 1,2, and 3 respectively.
    - Target Encoding: The categorical feature values are grouped by their corresponding target variable values. Some statistic (e.g. mean, median) of the target variable is calculated for each category group. This statistic is used to represent the category in the encoded data.
    - One-Hot Encoding (OHE): Creates a new binary feature for each category. Each feature is 1 for the corresponding category, and 0 for all others. Effective, but can lead to high dimensionality if there are many categories.
    - Hash Encoding: Categorical value is fed into the hash function. The hash function generates a unique (but potentially large) integer value. This integer value is then often mapped to a smaller range (e.g. 0 to number of features) to be usable by the model.

### Curse Of Dimensionality
- The curse of dimensionality refers to the challenges that arises when analyzing data with a high number of features (dimensions). As the number of dimensions increase, several issues can significantly impact the performance of ML algorithms and data analysis methods.
- Data sparsity:
    - Data points in low-dimensional space appear relatively closer and it is easy to find patterns, or draw relationships between them.
    - The same data points in high-dimensional space are spread out very thinly. Hence, the data becomes sparse, which makes it difficult to find patterns, or draw relationships between them.
- Increased distance metrics:
    - Distance metrics like Euclidean Distance becomes less informative in high dimensions. This can affect algorithms that rely on distance calculations, like KNN, for finding similar data points.
    - Therefore, Manhattan Distance is preferred when the number of dimensions increase.
- High computational cost: As the number of features increases, the computational cost of the comparison operations between 2 data points, or any operations on the data points grows exponentially.
- Overfitting: With a large number of features, there is a higher chance of random noise, or irrelevant features influencing the model. This can lead to overfitting.
- Impact:
    - On Machine Learning: Classification, Regression, Clustering algorithms can all suffer from the issues mentioned above, leading to poor performance and inaccurate results.
    - On Data Visualization: Visualizing high-dimensional data becomes challenging as it is difficult to represent many features in a single plot.

### Feature Importance
- Feature importance is a technique to determine the relative influence of input features on a model's predictions.
- Aids in:
    - Model interpretation.
    - Feature selection.
    - Prioritizing feature engineering.
- Feature importance in Linear Models:
    - Coefficient magnitude indicates feature importance.
    - Considerations:
        - Feature scaling: Unequal scaling can distort coefficient comparisons.
        - Feature correlation: Correlated features can obscure individual importance.
- Feature importance in Decision Trees:
    - Measured by Information Gain at each split.
    - Key considerations:
        - Data point distribution: Splits with more data points have higher Information Gain.
        - Feature re-occurrence: Importance is based on total Information Gain across all splits.

### Data Imbalance And Treatment
- Imbalance exists in dataset when data points belonging to a particular class are higher in number than the data points belonging to other class(es).
- Handling data imbalance:
    1. Oversampling techniques: These techniques increase the number of data points belonging to the minority class.
        - Random Oversampling (ROS): Duplicates data points from the minority class randomly.
        - SMOTE (Synthetic Minority Oversampling Technique): Creates synthetic data points for the minority class interpolating between existing minority class data points.
    2. Undersampling techniques: These techniques decrease the number of data points in the majority class.
        - Random Undersampling (RUS): Randomly removes data points from the majority class.
        - NearMiss (Nearest Neighbors): Removes majority class data points that are far away from minority class data points in feature space.
    3. Cost-Sensitive Learning/ Setting Class Weights: This approach assigns different costs to misclassifications of different classes. The model is trained to minimize the overall cost.

### Impact Of Data Balancing On The Model
- Positive impacts:
    - In imbalanced datasets, where 1 class has significantly greater number of data points (majority class) than the other (minority class), model tends to be biased towards the majority class. Balancing the dataset helps the model to learn the patterns of the minority class more efficiently, leading to improved performance.
    - Data balancing helps to mitigate the bias towards the majority class, leading to a fairer representation of both classes in the model's predictions. This is crucial for tasks where accurate identification of minority class is important.
    - Balancing techniques can sometimes lead to a model that generalizes better to unseen data, especially when dealing with imbalanced real-world scenarios.
- Negative impacts:
    - Oversampling techniques (duplicating minority class data) might lead to overfitting if not done carefully. The model might memorize noise or irrelevant patterns present in the duplicated data.
    - Oversampling the data increases training time and memory requirements, especially for large datasets.
    - In some cases, oversampling techniques can create artificial patterns in the minority class data, which might not reflect the real-world distribution.
    - Balancing techniques might not be very effective in cases of extreme class imbalance, where the minority class has very few data points.

# Evaluation Metrics Of Regression Models
### Error
- Error is the difference observed between the actual and the predicted values. Error is alternatively called as residual. Error is mathematically expressed as,
    - $\text{Error} = y_{actual} - y_{predicted}$.
- Ideally, error should be calculated by taking the sum of differences between actual values and predicted values ($\text{Error} = \sum_{i = 1}^n (y_{actual_i} - y_{predicted_i})$). But, the problem with this is that, if error were to be found in this way, there is a chance that the effect of positive values and negative values could cancel each other and the overall error may result in 0.
- Taking a mod is a good alternative. Except that, the derivative of mod is not defined at 0.

### Residual Sum Of Squares (RSS)
- RSS overcomes the above limitations. RSS is the sum of squares of errors. RSS is mathematically expressed as,
    - $\text{RSS} = \sum_{i = 1}^n (y_{actual_i} - y_{predicted_i})^2$.
- The drawback of using RSS as the error metric is that, it is dependent on the scale of the label. Therefore, the results may be exaggerated.

### Mean Squared Error (MSE)
- MSE calculates the average of sum of square of difference between the actual and the predicted values. A lower MSE indicates a that the Model fits on the data well. MSE is mathematically expressed as,
    - $\text{MSE} = \frac{1}{n} \sum_{i = 1}^n (y_{actual_i} - y_{predicted_i})^2$.

### Mean Absolute Error (MAE)
- MAE calculates the average of sum of mod of difference between the actual and the predicted values. MAE is less sensitive to outliers. A lower MAE indicates that the Model fits on the data well. MAE is not a preferred metric as the derivative of mod is not defined at 0. MAE is mathematically expressed as,
    - $\text{MAE} = \frac{1}{n} \sum_{i = 1}^n |(y_{actual_i} - y_{predicted_i})|$.

### Root Mean Square Error (RMSE)
- RMSE is the square root of average of sum of square of difference between the actual and the predicted values. RMSE is easier to interpret, as it is in the same units as the label. A lower RMSE indicates that the Model fits on the data well. RMSE is mathematically expressed as,
    - $\text{RMSE} = \sqrt{\frac{1}{n} \sum_{i = 1}^n (y_{actual_i} - y_{predicted_i})^2}$.

### R2 Score
- A better metric to evaluate a Model is by measuring it against an average Model. R2 Score implements this logic to calculate the error. It is mathematically experssed as,
    - $\text{R2 Score} = 1 - \frac{RSS}{TSS}$
    - RSS is defined as, $\text{RSS} = \sum_{i = 1}^n (y_{actual_i} - y_{predicted_i})^2$.
    - TSS (Total Sum of Squares) is defined as, $\text{TSS} = \sum_{i = 1}^n (y_{actual_i} - y_{average_i})^2$.
    - Although R2 Score is comparitively robust error metric, it does not take into account the number of features. 
    - Meaning, if there are $d$ features, and 1 relevant feature, the R2 Score should increase. Also, if there are $d$ features, and 1 irrelevant feature, the R2 Score should decrease. But in reality, the R2 Score remains the same in both the scenarios.

### Adjusted R2 Score
- The best metric to evaluate a Model is Adjusted R2 Score. Adjusted R2 Score takes in account the impact of number of features, and also the number of data points. It is mathematically expressed as,
    - $\text{Adjusted R2 Score} = 1 - \frac{((1 - \text{R2 Score}) * (n - 1))}{(n - d - 1)}$.
    - Where,
        - $n$ = Number of rows.
        - $d$ = Number of columns.

# Evaluation Metrics Of Classification Models
### Accuracy
- Accuracy is calculated using the following expression,
    - $\text{Accuracy} = \frac{\text{Number of Correct Prediction}}{\text{Total Number of Predictions}}$.
- Limitations of Accuracy being used as the evaluation metric:
    - Accuracy focuses on overall correctness. It does not provide insights into how well the model performs for each individual class.
    - In datasets where 1 class significantly outnumbers the others, the model can achieve a high Accuracy simply by predicting the majority class.

### Confusion Matrix
- Confusion Matrix is a visualization tool used to evaluate the performance of a classification model. It provides a clear breakdown of how the model performed on the testing data in terms of correctly and incorrectly classified data points for each class.
- Structure:
    | | Predicted Ham (Class 0) | Predicted Spam (Class 1) |
    | :-: | :-: | :-: |
    | Actual Ham (Class 0) | TN | FP |
    | Actual Spam (Class 1) | FN | TP |
- Accuracy can now be expressed as,
    $\text{Accuracy} = \frac{TP + TN}{TN + FP + FN + TP}$.

### Precision
- Precision measures the proportion of actual positives out of all the values that have been identified as positive.
- Precision is mathematically expressed as,
    - $\text{Precision} = \frac{TP}{TP + FP}$.

### Recall
- Recall measures the proportion of identified positives out of all the values that are actually positive.
- Recall is mathematically expressed as,
    - $\text{Recall} = \frac{TP}{TP + FN}$.

### F1 Score
- F1 Score is the harmonic mean of Precision and Recall. 
- F1 Score is mathematically expressed as,
    - $\text{F1 score} = \frac{2 * (\text{Precision} * \text{Recall})}{(\text{Precision} + \text{Recall})}$.
- Interpretation:
    - A high F1 Score (close to 1) indicates that the model is performing well on Precision and Recall.
    - An F1 Score of 0.5 indicates that the model is equal at Precision and Recall.
    - A low F1 Score (close to 0) indicates that the model is performing poorly at Precision and Recall.

### Sensitivity
- Sensitivity is also called as Recall or True Positive Rate (TPR). It measures the proportion of identified positives out of all the values that are actually positive.
- Sensitivity is mathematically expressed as,
    - $\text{Sensitivity} = \frac{TP}{TP + FN}$.

### Specificity
- Specificity is also called as True Negative Rate (TNR). It measures the proportion of identified negatives out of all the values that are actually negative.
- Specificity is mathematically expressed as,
    - $\text{Specificity} = \frac{TN}{TN + FP}$.

### False Negative Rate (FNR)
- FNR measures the proportions of actual positives that have been incorrectly identified as negative.
- FNR is mathematically expressed as,
    - $FNR = \frac{FN}{TP + FN}$.
- A high FNR is indicative of low Recall.
    
### False Positive Rate (FPR)
- FPR measures the proportion of actual negatives that have been incorrectly identified as positive.
- FPR is mathematically expressed as,
    - $FPR = \frac{FP}{TN + FP}$.
- A high FPR is indicative of low Precision.

### Receiver Operating Characteristics (ROC) Curve
- ROC Curves are a graphical tool used in ML to evaluate the performance of a Binary Classification Model. It illustrates the Model's ability to distinguish between positive and negative classes at various classification thresholds.
- ROC Curve is a plot with FPR (False Positive Rate) on the x-axis, and TPR (True Positive Rate) on the y-axis. A Model with higher Area Under ROC (AU-ROC) Curve generally performs better across all thresholds.
    - FPR represents the proportion of negative cases incorrectly classified as positive by the Model.
    - TPR represents the proportion of positive cases incorrectly classified as positive by the Model.
- Interpreting the AU-ROC Curve:
    - A perfect classifier would have an ROC Curve that goes straight up from the bottom left corner (0 FPR, 0 TPR) to the top left corner (0 FPR, 1 TPR), and then horizontally across the top (1 TPR for all remaining thresholds). This indicates that the model correctly identifies all positive cases without any False Positives.
    - A diagonal line from (0, 0) to (1, 1) represents a random classifier.
    - A curve closer to the top-left corner indicates better performance. This means the model is able to correctly classify more positive instances while minimizing False Positives.
    - The AU-ROC Curve provides a qunatitative measure of the model's performance. An AUC of 1 indicates perfect classification, while an AUC of 0.5 indicates random classification.
- ROC Curve can be misleading in cases where there is a severe class imbalance.

### Precision-Recall (PR) Curve
- PR Curves are a graphical tool used in ML to evaluate the performance of a Binary Classification Model. They are used when dealing with dataset where there is class imbalance. Unlike ROC Curves, PR Curves focus on the trade-off between Precision and Recall at various classification thresholds.
- PR Curve is a plot with Recall on the x-axis, and Precision on the y-axis.
- Interpreting the PR Curve:
    - A perfect classifier would have a PR Curve that starts at the top-left corner (0 Recall, 1 Precision) and goes diagonally down to the bottom right corner (1 Recall, 0 Precision). This indicates the model correctly identifies all positive cases without any False Positives.
- PR Curves can also be misleading in cases of severe class imbalance.

### Classification Report
- Classification Report is a summary of a Classification Model's performance on a dataset. It provides insights into how well the model classified data points belonging to different classes.
- Components of a Classification Report:
    - Classes: Different class labels present in the dataset.
    - Precision.
    - Recall.
    - F1 Score.
    - Support: Total number of data points belonging to each class in the dataset.
- Interpretation:
    - A good classification report will have high Precision, Recall, and F1 Score for all classes, ideally close to 1.
