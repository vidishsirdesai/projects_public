# Introduction
- In statistics, hypothesis testing is used to make inferences about the population based on a sample of data.
- The primary purpose of hypothesis testing is to determine whether a specific hypothesis about a population is supported by the sample data or not.
- Framework for hypothesis testing,
    1. Define H0 and H1.
    2. Set the $\alpha$ value.
    3. Choose the appropriate statistical test that is applicable.
    4. Find the test statistic and p-value.
    5. If p-value < $\alpha$, then reject H0. If p-value > $\alpha$, then accept H0.
- H0 or null hypothesis or default assumption suggests that there is no significant difference observed between existing data and new data for the same population parameter.
- H1 or alternative hypothesis suggests that there is a significant difference observed between existing data and new data for the same population parameter.
- During hypothesis testing, multiple alternative hypotheses can be formulated.
- The p-value helps in quantifying the strength of evidence against the null hypothesis. 
    - A large p-value suggests that the observed results are consistent with the null hypothesis.
    - A small p-value suggests that the observed results are different from the null hypothesis, supporting that the alternative hypothesis is true.
- Confidence Interval is the range of values that the estimate is expected to fall within, if the experiment is re-run or the population is reshaped in the same way.
- Significance Level is a pre-set threshold that is used to make decisions about the null hypothesis.
- $\text{Confidence Interval} = 1 - \alpha$.
- Types of errors:
    - Type 1 (False Positive).
    - Type 2 (False Negative).
- Types of hypothesis test:
    - Left tailed test: A left tailed test is carried out while determining if the sample statistic is lesser than a certain value.
    - Right tailed test: A right tailed test is carried out while determining if the sample statistic is greater than a certain value.
    - Two tailed test: A two tailed test is carried out while determining if the sample statistic is different from a certain value.

# Z-Test
- Z-Test is a statistical test used to determine if the sample mean is significantly different from population mean given that the population standard deviation, population mean, and sample mean are given.
- Framework:
    - Setup H0 and H1.
    - Find the sample standard deviation:
        - $\sigma_{sample} = \frac{\sigma_{population}}{\sqrt{n}}$.
        - Where,
            - $n$ = Sample size.
    - Find the Z-Score:
        - $\text{Z-Score} = \frac{\mu_{sample} - \mu_{population}}{\sigma_{sample}}$.
    - Find the p-value:
        - If right tailed, `p_value = 1 - norm.cdf(z_score)`.
        - If left tailed, `p_value = norm.cdf(z_score)`.
    - Compare the p-value with $\alpha$, and make a decision about the null hypothesis:
        - If p-value is lesser than $\alpha$, then reject H0.
        - If p-value is greater than $\alpha$, then accept H0.

# T-Test
- T-Test is a statistical test which is very similar to Z-Test. It is used to determine if there is significant difference between the means of 2 groups or if a single group's mean is different from a known value.
- Z-Test is used when the population standard deviation is known, where as T-Test is used when the population standard deviation is unknown.
- Z-Test is used while working with a large sample size, where as T-Test is used while working with a relatively smaller sample size.
- Type of T-Test:
    - 1 Sample T-test: Used to test if the mean of a single sample is significantly different from a known or hypothesized population mean. Types:
        - Right tailed.
        - Left tailed.
    - 2 Sample T-Test (Independent Sample T-Test): Used to compare the means of 2 independent groups to assess if they are significantly different from each other. Types:
        - Right tailed.
        - Left tailed.
    - Paired Sample T-Test: Used when dealing with paired or matched samples, such as before and after measurements of the same group.

# Chi-Squared Test
- Degrees-of-Freedom (DoF) are the bare minimum numbers in a dataset that are required to be known in order to reconstruct the dataset. For a dataset containing n elements, (n - 1) elements are required to be known in order to find the 1 unknown.
- Chi-Squared Test is a statistical test used to determine whether there is a significant association between 2 categorical variables. Types:
    - Test for independence: Used to examine whether there is an association between 2 categorical variables. It determines if the variables are independent of each other or if they are related.
    - Test for goodness of fit: Used to determine if the observed frequency distribution of a single categorical variable differs from an expected frequency distribution. In simpler terms, it checks if the observed data fits a particular expected pattern.
- Assumptions of Chi-Squared Test:
    1. Variables are categorical.
    2. Observations are independent.
    3. Each cell is mutually exclusive.
    4. The expected value in each cell has to be greater than 5.

# ANOVA (Analysis Of Variance)
- ANOVA is a statistical method used to compare the means among 3 or more groups.
- It assesses whether there are any statistically significant differences in the means of a dependent variable across different groups.
- Assumptions of ANOVA:
    1. Data should be Gaussian (a QQ-plot can be plotted or Shapiro Test helps in confirming this).
    2. Data is independent of each other.
    3. Equal variances in different groups (Levene Test helps in confirming this).

# Kruskal-Wallis Test
- Kruskal-Wallis Test can be conducted to compare the means among 3 or more groups in case the assumptions of ANOVA are not satisfied.

# Covariance
- Covariance is the average of all variances.
- It is expressed as,
    - $\text{Cov}(H, W) = \frac{1}{n} \sum_{i = 1}^n (H_i - \bar{H}) (W_i - \bar{W})$. Where,
        - $\text{Cov}(H, W)$ = Covariance between $H$ and $W$.
        - $H_i$ = i-th value of the variable $H$.
        - $\bar{H}$ = Mean of the variable $H$.
        - $W_i$ = i-th value of the variable $W$.
        - $\bar{W}$ = Mean of the variable $W$.
        - $n$ = Total number of data points.
- The problem with Covariance is that it depends on the scale of the entities involved. This problem is addressed by dividing the Covariance by standard deviation.
    - $\text{Cov}(X, Y) = \frac{1}{n} \sum_{i = 1}^n \frac{(H_i - \bar{H})}{\sigma_h} \frac{(W_i - \bar{W})}{\sigma_w}$.
- This equation is termed as Correlation Coefficient, and is represented by $\rho$.
    - $\rho = \frac{1}{n} \sum_{i = 1}^n \frac{(H_i - \bar{H})}{\sigma_h} \frac{(W_i - \bar{W})}{\sigma_w}$.
    - If the $\rho$ is close to 0, then it is said that the data has no correlation.
    - If the $\rho$ is close to 1, then it is said that the data has strong positive correlation.
    - If the $\rho$ is close to -1, then it is said that the data has strong negative correlation.

# Pearson Correlation Test
- It measures the linear relationship between 2 continuous variables.
- It assumes that the data are normally distributed and that there is a linear relationship between the variables.

# Spearman Rank Correlation Test
- It measures the monotonic relationship between 2 variables, whether its linear or not.
- It does not assume a linear relationship and is suitable for ordinal, or non-normally distributed data.

# Kendall's Tau Rank Correlation Test
- It measures the ordinal association between 2 variables.
- It does not assume a linear relationship and is suitable for ordinal or non-normally distributed data.
