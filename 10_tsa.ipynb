# Introduction
Forecasting is predicting future trends based on past data, and potential known, or anticipated future events. The potential future events could be general elections, wars, global crisis, or local crisis, or any controversies that surround the event that is being predicted.

Forecasting as an ML technique falls under the broader category of supervised ML.

Forecasting works in the following way,
- A dataset with a time parameter $t$, and historic data, say a variable $x$ with values $x_1$, $x_2$, $x_3$, ..., $x_t$ is considered.
- For this variable $x$, the future values, represented by $x_{t+1}$, $x_{t+2}$, ..., $x_{t+k}$, are predicted using the past values.

Forecasting can be performed for a set of future values, this set of future values can be a few days, weeks, months, quarters, or years in the future. It depends on the problem statement and the requirement.

### Linear Regression vs. Forecasting
- In both the cases, a variable of numerical data type is predicted.
- Time does not play a role in Linear Regression. Whereas, time plays a rather important role in Forecasting, because it is expected that the forecast be closer to recent data than older data.
- A Time Series Forecasting Model usually has 2 or sometimes more features as inputs. The input is time or date, and historical data of the variable that is being predicted, and the output will be the future values of the variable that is being predicted. On the other hand, a Linear Regression Model can have many inputs.

# Handling Missing Values
The best practice is to not drop missing values in Time Series Analysis. This is because, there will be components of seasonality in the data, and the prediction of future values depends on this seasonality. Hence, the element of time is very important for prediction, missing even 1 value will result in missing out on the accuracy of the future values.

### Imputation Techniques
- Start by finding out the missing values in the dataset.
- The mean value, or the median value of the dataset can be used to impute the missing values. 0 can also be used to impute the missing values.
- A better method to impute missing values is using the interpolation technique.
- In interpolation technique, the mean of $n$ values before, and $n$ values after the missing value is used to impute the missing value. This technique is called as centered Moving Average.

# Outliers Treatment
Outliers can be handled using either one of the following two techniques:
- Trimming:
    - Trimming involves removing a certain percentage of data from both tails of the distribution.
    - For example, if 5% is trimmed from each tail of the dataset, the lowest, and the highest 5% of the data will be removed.
    - Trimming helps to reduce the influence of extreme values on statistical estimates and can make the analysis more robust to outliers.
    - However, trimming may lead to loss of information, particularly if the extreme values contain important insights or represent genuine observations.
- Winsorizing:
    - Winsorizing is a method that replaces extreme values (outliers) with less extreme values, typically by setting them to the value at a specified percentile.
    - For example, if winsorization is performed at 95th percentile, any value below 2.5th percentile are set to the value at the 2.5th percentile, and any value above the 97.5th percentile are set to the value at the 97.5th percentile.
    - Winsorizing retains the shape of the distribution while reducing the impact of the outliers.
    - It is less extreme than trimming because it does not completely remove the outlier values but rather moderates their influence.

Since data entries cannot be dropped in Time Series Analysis, Winsorizing is the goto technique to deal with the outliers. The practice is to floor at the 5th percentile and cap at 95th percentile. But this can be changed as per the requirement and the final objective.

# Components In A Time Series
A Time Series has the following 4 components.
1. Level: Level is defined as the base value on which the forecast is built upon.
2. Trend: 
    - Trend is the direction (upwards or downwards) in which the Time Series is growing in.
    - The data can have one of the following trends,
        1. Up Trend.
        2. Down Trend.
        3. Changing Trend.
        4. Flat Trend.
    - Generally, the best way to plot a Trend is to consider a long-term Moving Average.
3. Seasonality:
    - Seasonality is the repeated pattern that is observed in the plot of the dataset. The repetition can be daily weekly, monthly, quarterly, etc.
    - Seasonality in Time Series refers to a pattern that occurs at a regular interval.
    - Seasonality usually repeats after a fixed duration of time. In fact, the assumption that a Time series can be predicted is due to the fact tht Seasonality is repeatitive.
4. Error: 
    - Error represents the random, irregular fluctuations that are not explained by the Level, Trend, or Seasonality. It is essentially the residual, or the noise left over after accounting for these systematic components.
    - The Error component in Time Series, like in Linear Regression, needs to be normally distributed. The mean of Error should be close to 0.
    - Generally when the Error in Time Series does not have a normal distribution, and the mean of Errors is not equal to 0, it means that the Error also has some Time Series (Error is also increasing, or decreasing with time). Further decomposition of Error should be done until it becomes normally distributed.

### Mathematical Representation of Time Series
$\text{Time Series} = y(t) = \text{Level} + \text{Trend}(t) + \text{Seasonality}(t) + \text{Error}$.

$y(t) = l + b(t) + s(t) + e(t)$.

Where,
- $l$ = Level of Time Series.
- $b(t)$ = Trend of Time Series.
- $s(t)$ = Seasonality of Time Series.
- $e(t)$ = Error of Time Series.

Level does not have a time component associated with it.

The equation of linear regression is given by,

$y = (w^T * x) + e$.

Where,
- $w$ = Weight.
- $x$ = Independent variable.
- $e$ = Error.

The trend will be the slope of this line.

Time series is almost similar to the equation of Linear Regression, if an observation is made.

# Types Of Time Series
Time Series can be decomposed into the following 2 types,
1. Additive Decomposition:
    - $y(t) = l + b(t) + s(t) + e(t)$.
2. Multiplicative Decomposition:
    - $y(t) = l * b(t) * s(t) * e(t)$.

The peaks and troughs in Multiplicative Decomposition are higher and lower, which is opposite in Additive Decomposition.

# Mean Absolute Percentage Error (MAPE)
MAPE is the average of sum of the ratio of mod of difference between the the actual value and the predicted value and the actual value. It is mathematically expressed as,
- $\text{MAPE} = (\frac{1}{n} \sum_{i = 1}^n \frac{|y_{\text{actual}} - y_{\text{predicted}}|}{y_{\text{actual}}}) \times 100$.

There are other metrics like, MSE, RMSE, MAE, etc, which can be used to evaluate the performance. MAPE is the preferred choice of evaluation metric in case of Time Series, because it gives an idea of the absolute performance. Essentially, MAPE defines how off is the predicted value from the actual value. For example, if MAPE = 37.5%, then the predicted value is off by 37.5% from the actual value.

In comparison, RMSE gives the idea about relative performance. Meaning, it only gives a value and does not define what the value represents. Meaning, if a Model M1 has an RMSE of 15, and another model M2 has an RMSE of 25, then M1 is better than M2 for sure, but it does convey any information about how M1 is better than M2.

MAPE cannot be calculated when $y_{\text{actual}} = 0$. The acceptable range of MAPE varies from industry to industry, or from objective to objective.

# Train-Test Split
In Linear Regression, the split is done such that, train dataset is assigned 80% of the data, and the test dataset is assigned with the remaining 20% of the data.

In Time Series Analysis, on the contrary, random train-test split cannot be done the way it is done for Linear Regression, or other ML models. This is because, the order of time is very important while Forecasting. Therefore, if there are 200 points, and if the split is to be done such that train dataset has 70% of the data, and test dataset has 30% of the data, the first 140 points are assigned to the train dataset, and the remaining 60 points are assigned to the test dataset.

Because the sequence of datetime column should not be altered in Time Seies Analysis, random splitting of the dataset into train and test should be avoided. Hence, the split is done by slicing.

Things to keep in mind,
- Indexing is done on date.
- Train-Test split is not random.
- Latest data will go into test.

# Simple Forecasting Techniques
### Mean Forecasting
Consider a plot of time versus sales, and prediction has to be made for $t+1$, $t+2$, $t+3$, ... and so on. In order to do so, the mean of the entire data points is calculated, and this calculated mean value is considered as the forecast for $t+1$.

This procedure is repeated for $t+2$, $t+3$, ... (the previous value, i.e., $t+1$ is considered as well while forecasting for $t+2$).

This very simple forecasting technique is generally used while making a short term prediction. Trend and Seasonality is not captured in Mean Forecasting.

Note that, adding a mean value to the dataset, does not change the mean of the dataset.

### Naive Forecasting
In Naive Forecasting, the last or the latest value is used as the forecast for the next value. Meaning, if a forecast is to be made for time = $t$, then the vaue at $t-1$ is used as the forecast for $t$.

Trend and Seasonality is not captured in Naive Forecasting. The only difference between Naive Forecasting, and Mean Forecasting is that, more weightage is assigned to the recent data.

### Seasonal Naive Forecasting
In Naive Forecasting, all the future forecasts will have the same value. In Seasonal Naive Forecasting, the forecast for Jan 2023 will be the same value as the value for Jan 2022. Hence, an element of seasonality is captured to some extent in this Forecasting technique.

### Drift Forecasting
Drift Forecasting makes use of the concept of slope to bake the trend in the Time Series to predict the future values.

Consider a plot of time versus sales, say that there is an increasing trend observed. If this is the case, it is generally expected that, in the forecast, the trend will be increasing as well.

The equation of slope is given by, $m = \frac{(y_t - y_0)}{(t - t_0)}$.

Consider, $y = mx + c$. And, $m = \frac{(y_{t+h} - y_t)}{((t + h) - t)}$.

Therefore, $y_{t+h} = y_t + m \times h$.

Where,
- $y_t$ = Latest value in the train set.
- $m$ = Slope.
- $h$ = Number of steps being predicted in the future.

The assumption is that if there is an increasing trend observed from $y_0$ to $y_t$, then the trend should continue into $y_{t+1}$, $y_{t+2}$, and so on.

The value $y_{t+1}$ is calculated using,
- $y_{t+1} = y_t + m \times 1$.

Similarly, the value $y_{t+2}$ is calculated using,
- $y_{t+2} = y_t + m \times 2$.

And so on.

Drift Forecasting captures the trend, but not seasonality. In order to get a better prediction, capturing seasonality is more important than trend.

### Forecasting Using Moving Averages
In Forecasting using Moving Average, the mean of last few entries is considered to forecast the next value. For example, if Forecasting for Jan 2023, the mean of data from Jan 2022 to Dec 2022 is computed and used as the forecast for Jan 2023. Similarly, if Forecasting for Feb 2023, the mean of data from Feb 2022 to Jan 2023 is computed and used as the forecast for Feb 2023.

Both trend and seasonality are captured to a good extent in this technique.

### NOTE
All the simple Forecasting techniqes are deployed when something quick and hacky is required. These techniques generate forecasts that have very low fidelity.

# Forecasting Using Smoothing Based Models
Consider a plot of time vs. sales, where the trend is flat-drop-flat-drop-flat-drop. Divide into 3 windows, $a$, $b$ and $c$ each one representing a flat-drop trend. While making a prediction, the most important among them is the window $c$.

The prediction for $t + 1$ is mathematically expressed as,
- $y_{t+1} = \alpha * y_t + (1 - \alpha) * y_{t-1}$.
- Where,
    - $y_{t+1}$ = Future Prediction.
    - $y_t$ = Latest Value.
    - $y_{t-1}$ = (Latest Value - 1).

Meaning, a weightage of $\alpha$ is given to the latest value, and a weightage of $(1 - \alpha)$ is given to (latest - 1) value.

While calculating moving average, some or all the data points from the past if considered for computing the average. The problem is that, when a flat average is taken, all the future forecasts will show no trend or seasonality.

To solve this problem, weightage is assigned to the data points. Most recent data point is assigned the highest weightage, and the weightage assigned to the data point is decreased as the data points get older.

Generally, if there are $n$ data points, the weights, $α_1$, $α_2$, $α_3$, ..., $α_n$ are assigned to them. The weights assigned to them are such that, sum of weights assigned to data points when added will be equal to 1. Meaning, 
- $\sum{\alpha_{i}} = \alpha_1 + \alpha_2 + \alpha_3 + ... + \alpha_n = 1$.

Therefore, $\sum{\alpha_i} = 1$.

The plot of time vs weight (weights assigned) is such that as the time decreases, the weight decreases exponentially. The overall sum would be equal to 1.

This is the underlying principle behind Forecasting using smoothing techniques.

Consider the following example dataset,

| month | sales |
| :-: | :-: |
| Jan 2023 | 100 |
| Feb 2023 | 150 |
| Mar 2023 | 160 |
| Apr 2023 | 180 |
| May 2023 | 200 |
| Jun 2023 | 250 |
| Jul 2023 | np.nan |

If a prediction is to be made for Jul 2023, the weights would be assigned in the following fashion,

| month | sales | weightage |
| :-: | :-: | :-: |
| Jan 2023 | 100 | 0.05 |
| Feb 2023 | 150 | 0.07 |
| Mar 2023 | 160 | 0.07 |
| Apr 2023 | 180 | 0.1 |
| May 2023 | 200 | 0.3 |
| Jun 2023 | 250 | 0.5 |
| Jul 2023 | np.nan | np.nan |

The total weightage assigned = 1.

The prediction for Jul 2023 is mathematically given by,
- $\text{Jul 2023} = (250 * 0.5) + (200 * 0.3) + (180 * 0.1) + (160 * 0.07) + (150 * 0.07) + (100 * 0.05)$.

The Time Series is made up of the following components,
- $\text{Time Series} = \text{Level} + \text{Trend} + \text{Seasonality} + \text{Error}$.
- $y(t) = l + b(t) + s(t) + e(t)$.

Level is the base value at which the forecasting starts.

Consider the following Time Series,
1. A Time Series with flat-drop-flat-drop-flat-drop line. This is clearly a declining trend. But if a small window is analyzed, say a flat, there is no trend observed. Also there is no seasonality observed for the same.
2. In a Time Series with increasing wavy line with small peaks and troughs, a trend is observed, but there is no seasonality.
3. In a Time Series with increasing wavy line with large peaks and troughs, a trend and seasonality both are observed.

All of the above 1, 2, and 3 are examples of Time Series, of course with noticeable differences between them.

- To perform any forecasting using the 1st chart, only level and error can be considered, because there is no trend and seasonality. That is, $y(t) = \text{Level} + \text{Error}$.
- To perform any forecasting using the 2nd chart, level, trend and error can be considered. That is, $y(t) = \text{Level} + \text{Trend} + \text{Error}$.
- To perform any forecasting using the 3rd chart, level, trend, seasonality and error can be considered. That is, $y(t) = \text{Level} + \text{Trend} + \text{Seasonality} + \text{Error}$.

### Simple Exponential Smoothing
Simple Exponential Smoothing (SES), is used when the Time Series has no trend and no seasonality (Time Series 1). In order to predict for $t+1$, weights are assigned to the data points, highest weight is assigned to the most recent data point. As the recency decreases, the weight value assigned also decreases. The equation looks as follows,
- $y_{t+h} = (\alpha * y_t) + (\alpha(1 - \alpha) * y_{t-1}) + (\alpha(1 - \alpha)^2 * y_{t-2}) + (\alpha(1 - \alpha)^3 * y_{t-3}) + ... + (\alpha(1 - \alpha)^n * y_{t-n})$.
- Where,
    - $\alpha$ = Smoothing of Level.

Weight here is a hyperparameter. Generally,
- $\alpha = \frac{1}{2 * Seasonality}$.

If the seasonality is repeating after every 12 months, then,
- $\alpha = \frac{1}{2 * 12} = \frac{1}{24}$

The above will be the starting value of $\alpha$. But since it is a hyperparameter, it is open for exploration.

### Double Exponential Smoothing (Holt Method)
Double Exponential Smoothing (DES), is used when the Time Series has a trend and no seasonality (Time Series 2). In this technique, dataset which has Time Series 2 is considered. The Time Series is represented as,
- $y(t) = \text{Level} + \text{Trend} + \text{Error}$.

In order to predict for $t+h$,
- $y_{t+h} = ((\alpha * y_t) + (\alpha(1 - \alpha) * y_{t-1}) + ... + (\alpha(1 - \alpha)^n * y_{t-n})) + ((\beta * b_t) + (\beta(1 - \beta) * b_{t-1}) + ... + (\beta(1 - \beta)^n * b_{t-n}))$.
- Where,
    - $\alpha$ = Smoothing of Level.
    - $\beta$ = Smoothing of Trend.

### Triple Exponential Smoothing (Holt-Winter's Method)
Triple Exponential Smoothing (TES), is used when the Time Series has a trend and a seasonality (Time Series 3). In this technique, dataset which has Time Series 3 is considered. The Time Series is represented as,
- $\bar{y_{t+h}} = \text{Level} + \text{Trend} + \text{Seasonality} + \text{Error}$.

In order to predict for $t+h$,
- $y_{t+h} = ((\alpha * y_t) + (\alpha(1 - \alpha) * y_{t-1}) + ... + (\alpha(1 - \alpha)^n * y_{t-n})) + ((\beta * b_t) + (\beta(1 - \beta) * b_{t-1}) + ... + (\beta(1 - \beta)^n * b_{t-n})) + ((\gamma * s_t) + (\gamma(1 - \gamma) * s_{t-1}) + ... + (\gamma(1 - \gamma)^n * s_{t-n}))$.
- Where,
    - $\alpha$ = Smoothing of Level.
    - $\beta$ = Smoothing of Trend.
    - $\gamma$ = Smoothing of Seasonality.

### NOTE
The method `statsmodels.tsa.ExponentialSmoothing()` is used to deal with smoothing of trend or seasonality. Where as `statsmodels.tsa.SimpleExpSmoothing()` is used while dealing with just level.

# Auto Correlation Factor And Partial Auto Correlation Factor
### Auto Correlation Factor (ACF)
- ACF the correlation between a Time Series, and its own lagged (previous) values. In simple words, it defines how much the current value of a Time Series is related to its values at previous time steps.
- The lag ($k$) represents the number of time periods between the current value, and the past value being compared. For example, a lag of 1 compares the current value with the value from the immediately preceding time period, a lag of 2 compares with the value from two periods ago, and so on.
- The ACF at lag $k$, denoted as $\rho_k$, is essentially, the Pearson Correlation Coefficient between the Time Series $y_t$, and its lagged version $y_{t - k}$.
    - $\rho_k = \text{Corr}(y_t, t_{t - k})$.
- Interpretation of ACF Plot (Correlogram): The ACF plot shows the correlation coefficient on the y-axis and the lag on the x-axis.
    - Randomness: For a purely random Time Series (white noise), the autocorrelations at all lags (except lag 0, which is always 1) should be close to zero.   
    - Stationarity: For a stationary Time Series, the ACF typically decays gradually as the lag increases.
    - Trends: Time Series with trends often exhibit high positive autocorrelation at small lags, which gradually decreases as the lag increases.   
    - Seasonality: Seasonal patterns in a Time Series are often reflected in the ACF plot as significant spikes at lags corresponding to the seasonal period and its multiples.

### Partial Auto Correlation Factor (PACF)
- PACF measures the direct correlation between a Time Series and its lagged values, after removing the effect of the intermediate lags.
- PACF helps to isolate the direct impact of a specific lagged value on the current value.
- Interpretation of PACF Plot:
    - The PACF plot shows the partial correlation coefficient on the y-axis and the lag on the x-axis.
    - The PACF is particularly useful for determining the order of an AR model.   
    - For an $\text{AR}(p)$ process, the PACF will have significant spikes up to lag $p$ and then drop sharply to zero (or become statistically insignificant) for lags greater than $p$.

# Stationarity
Stationarity in Time Series is a condition where, parameters like, mean, variance, amplitude, frequency do not change naturaly with time. Essentially, the Time Series will not have any trend or seasonality in it.

Consider a Time Series wth increasing wavy line with large peaks and troughs. This Time Series can be broken down into the following parts,
- Level.
- Trend.
- Seasonality.
- Residuals (Errors).

In order to use ARIMA based models for Forecasting, the Time Series should be rid of trend and seasonality, and only error should be retained.

### Dickey-Fuller Test
To confirm if a Time Series is stationary or not, Dickey-Fuller Test can be conducted. If the p-value obtained is greater than 0.05, the Time Series is considered not stationary (has trend and seasonality). The H0 and H1 for the test are,
- H0: Time Series is not stationary.
- H1: Time Series is stationary.

### Converting Non-Stationary Time Series To Stationary Time Series
- Removing Trend (De-trending):
    - Usually, the trend can be removed from the Time Series by differentiating once.
    - Or, the Time Series is differentiated until the trend is removed.
- Removing Seasonality (De-seasonalizing):
    - Seasonality can be removed by differentiating the Time Series by differentiating by seasonality found using ACF and PACF.

### NOTE
- If the Time Series cannot be converted into being stationary, then ARIMA based models are not used.

# Correlation Versus Causation
### If a variable x is correlated with another variable y, does that mean x causes y?
A variable x may be useful for Forecasting a variable, but that does not mean x is causing y. For example,
- It is possible to model the number of drownings at the beach resort each month with the number of ice-creams sold in the same period.
- The model can give reasonable forecasts, not because ice-creams cause drownings, but because people eat more ice cream on hot days when they are also more likely to go swimming at the beach.

So the 2 variables, ice-cream and drowning are correlated, but one is not causing the other.

They are both caused by a 3rd variable, i.e., temperature.

This is an example of confounding.

### What is the relevance of correlation and or causality with Forecasting Models?
Correlations are useful for Forecasting, even when there is no causal relationship between the two variables. For example,
- It is possible to forecast if it will rain in the afternoon by observing the number of cyclists on the road in the morning.
- When there are fewer cyclists than usual, it is more likely to rain later in the day.

This model can give reasonable forecasts, not because cyclists prevent rain, but becaue people are more likely to cycle when there is less or no chance of rain.

In this case, there is a causal relationship, but in the opposite direction to the Forecasting model.

### If good forecasts can be based on correlated variables, why bother trying to understand the causality behind them?
Though a good forecast can be obtained based on correlated variables, if the causality behind those variables is understood, better features can be identified, thereby creating a better model.
- A better model for drownings example will probably include temperature, and visitor numbers, and exclude ice-cream sales.
- A good Forecasting Model for rainfall will not include cyclists, but it will include atmospheric observations from the previous few days.

# Confidence Interval
Confidence interval is the mean of the estimate +/ - the explained variance.

Consider that the average height of the Indian males has to be calculated. The approaches that can be taken are,

Measure the height of all the Indians. The mean value obtained after the heights of all the Indians are measured is called as the population mean. This would also be the true average of the Indian males. But the reality of doing this is that, India is a country with a population of 1.4 billions roughly, and 52% of the population are males, meaning 0.74 billions. Going out there and measuring the heights of each of these individuas is a tedious task.

But is a small sample out of this 0.74 billion is taken, say 1 ;akh people (the sample can be random or stratified). This sample is treated as a true representation of the population entirely.
- Say that the mean height of this sample is 5ft7in. This mean is called as the sample mean.
- Say that another random population of 1 lakh is considered, and the sample mean for this is found to be 5ft6in.
- A hundred or thousand such samples can be considered, and each of them will have a sample mean.
    
This variation in sample mean happens because of sampling error.

Therefore, rather than quoting an absolute value, a range is suggested. It is usually said in the following way,

"Based on the 100 or 1000 samples, the average height of Inidan males is between 5ft6in to 5ft9in, and there is a 95% chance that the height can lie in this range."

This is called as confidence interval (CI).

### Confidence Interval In Forecasting
In forecasting, a prediction is performed. Consider a time series dat which has an increasing wavy pattern. Say that some future value is being predicted. Would this predicted value be accurate? Since this is a prediction, there is always a probability of error, and hence it will not be accurate. Therefore, a confidence interval is calculated, in addition to the predicted value. This confidence interval abviously has a range. Therefore, there are 3 values, predicted value, upper interval of the range, and the lower interval of the range.

# Exogenous Variable
Exogenous quiet simply means outlier or external or alien.

Consider a dataset of date and traffic between Bangalore and Mysore. When plotted, there are sudden surges seen. One of the possible reasons could be a long weekend, where people are travellling to nearby hill stations like Ooty, Coonoor, Kodaikanal, etc. Long weekends don;t occur at the same time every year. Such scenarios, which are not predictable are called exogenous variable. Such events cannot be planned for, and hence corrective measures can be taken when they occur.

In forecasting there is something called as momentum forecast. Momentum forecast is basically done based on historical data. But exogenous variables are unknown and hence they cannot be factored in.

Another example is, rise in the internet usage during the COVID-19 pandemic.

The changes observed during the normal weekends is not exogenous in nature.

Real world events like World Cups/ Olympics/ Global Events/ Big Movie releases etc, are all examples of exogenous events.

Examples of exogenous events:
- If you are working in Air-Conditioning industry, sales usually increase from April to June, but sometimes the sales also go up during January to February as well, this is due to El Niño effect. This is also an example of exogenous event.
- Say if the pakoda/ samosa consumption increases out of nowhere, due to rainfall. This is also an example of exogenous event.
- Oil prices going up due to a war is also an example of exogenous event.

Therefore, any event that has not been witnessed periodically, and if such events mess up the real world forecasts, those events are called exogenous events. In a nutshell, anything that is unanticipated and leads to a change in the forecast is called exogenous variable.
