# Defining The Classification Problem Mathematically
### Dataset
$D : \{(x_i, y_i): x_i \in \Re^d, y_i \in \{-1, 1\}\}^n_{i = 1}$

Where,
- $x_i$ = Features.
- $y_i$ = Input labels.
- $R^d$ = d-Dimension real numbers (size of the $w$ vector).

### Goal
Find a linear or non-linear function f(x) such that,

$f(\bar{x}) = y_i'$. Where,
- $\bar{x}$ = New vector.
- $y_i'$ = Prediction.

If the model is good, then $y_i'$ = $y_i$ (ideally).

Gain function is defined as,

$G(D, w, w_0)$

$G = \frac{1}{n} \sum^n_{i = 1} \frac{\overrightarrow{w}^Tx_i + w_0}{||w||} * y_i$

Converting the above to a different format,

$G = \sum^n_{i = 1} g(x_i, y_i, \overrightarrow{w}, w_0)$

The $g$ in the above equation represents the distance of a single point from the line represented by $\overrightarrow{w}$ and $w_0$. And $i = 1$ to $n$ is the summation of all the distances.

### Why summation and not multiplication?
- $\sum$ = Summation.
- $\Pi$ = Multiplication.

Even if one point is present on the line, the distance of it will be zero and hence, the whole operation will result in zero.

Hence summation is preferred over multiplication. Log or average is also preferred sometimes.

$g(x_i, y_i, \overrightarrow{w}, w_0) = \frac{\overrightarrow{w}^Tx_i + w_0}{||w||} * y_i$.

# Using Derivatives For Optimization
1. Slopes and critical points:
    - Slope as derivative: The slope of a curve at a point is given by the first derivative of the function at that point, denoted as $f'(x)$.
    - Critical points: Points where the slope is zero, i.e., $f'(x) = 0$, are called critical points. These points are potential candidates for maxima, minima, or saddle points.
    - Visualizing slopes: A negative slope means the curve is decreasing (angle > 90˚ with x-axis), and a positive slope means the curve is increasing (angle < 90˚ with the x-axis).

    

2. Minima and maxima:
    - Minima (Valleys):
        - At a minimum point, the slope changes from negative to positive.
        - The function reaches a local minima value.
    - Maxima (Peaks):
        - At a maximum point, the slope changes from positive to negative.
        - The function reaches a local maximum value.
3. Finding candidate points:
    - First derivative:
        - The first step is to find all points where the first derivative $f'(x) = 0$.
        - This gives the critical points.
    - Segregation challenge:
        - However, this condition alone does not tell whether a critical point is a maximum, minimum, or saddle point.
        - To deal with this, derivative of a point just before and after the point where $f'(x) = 0$ can be taken.
        - This is not the preferred way, as it is not very efficient.
4. Second derivative test:
    - Second derivative:
        - The second derivative, $f''(x)$, tells about the concavity of the function.
        - It indicates the rate of change of the slope.
    - Test rules:
        - $f''(x) > 0$ (Concave up): The function is concave up, and the critical point is local minimum. The rate of change of derivative is positive. Meaning, the slope is increasing.
        - $f''(x) < 0$ (Concave down): The function is concave down, and the critical point is a local maximum. The rate of change of derivative is negative. Meaning the slope is decreasing.
        - $f''(x) = 0$: The test is inconclusive. The critical point could be a saddle point, a minimum or a maximum. Further analysis is needed.
    - Intuition:
        - A positive second derivative means the slope is increasing, indicating a valley (minimum).
        - A negative second derivative means the slope is decreasing, indicating a peak (maximum).
5. Saddle points:
    - Definition:
        - A saddle point is a critical point where the function is neither a maximum nor a minimum.
        - It resembles a saddle shape, where the function increases in one direction and decreases in another.
    - No change in slope:
        - There is no consistent change in the slope before and after a saddle point.
        - The second derivative test is inconclusive at saddle points.
    - Example:
        - The function $f(x, y) = x^2 - y^2$ has a saddle point at $(0, 0)$.
6. Optimization applications:
    - Finding optimal parameters:
        - In ML and optimization problem, often, there is need to find the values of parameters that minimize or maximize a function (e.g., a loss function or a gain function). Derivatives are essential for this process.
    - Gradient descent:
        - Gradient descent is an iterative optimization algorithm that uses the first derivative to find the minimum of a function.
        - It repeateadly updates the parameters in the direction of the negative gradient (the direction of the steepest decrease).
    - Second-order methods:
        - Second-order methods, such as Newton's method, use the second derivative to find the minimum or maximum of a function more efficiently.
        - These methods often converge faster than gradient descent.

# Example
Consider the function, $f(x) = x^3 - 3x^2 + 1$.

1. Find the first derivative:
    - $f'(x) = \frac{d}{dx}(x^3 - 3x^2 + 1) = 3x^2 - 6x$.
2. Find the critical points by setting the first derivative to zero:
    - $f'(x) = 3x^2 - 6x = 0$.
    - $3x(x - 2) = 0$.
    - This gives 2 critical points:
        - $x = 0$.
        - $x = 2$.
    - At these points, the first derivative is indeed 0.
3. Find the second derivative:
    - Differentiating the first derivative $f'(x) = 3x^2 - 6x$ to find the second derivative $f''(x)$:
        - $f''(x) = \frac{d}{dx}(3x^2 - 6) = (6x - 6)$.
4. Evaluate the second derivative at the critical points:
    - For $x = 0$:
        - $f''(0) = 6(0) - 6 = -6$.
        - Since $f''(0) = -6 < 0$, the second derivative test tells that there is a local maximum at $x = 0$.
    - For $x = 2$:
        - $f''(0) = 6(2) - 6 = 6$.
        - Since $f''(0) = 6 > 0$, the second derivative test tells that there is a local minimum at $x = 2$.

# Working Of The Gradient Descent Algorithm
1. The hill analogy:
    - Consider a man standing on top of a hill, who wants to come down to the bottom of the hill. Also consider that, the visibility is low due to fog and he has got no one to help.
    - The problem:
        - The man on the hill represents an optimization problem where there is a need to find the minimum of a function (the bottom of the hill).
        - The fog represents the lack of complete information about the function's landscape.
    - The solution:
        - Small steps: Taking small steps ensures that the man does not overshoot the minimum or get stuck in a local minimum. This corresponds to using a small learning rate in gradient descent.
        - Downhill movement: Moving downhill means following the direction of the steepest descent, which corresponds to the negative gradient of the function.
        - Iteration: The man continues this process until he reaches the bottom, just as gradient descent iterates until it converges to a minimum.
2. Mathematical example - $f(x) = x^2$:

    <img src = "../artifacts/linear_algebra_and_optimization_65.png" width = "300" height = "400">

    - Function and goal:
        - The goal is to find the minimum of the function.
        - In this case, it is known that the minimum occurs at $x = 0$.
    - Gradient (Slope):
        - The derivative (slope) of $f(x) = x^2$ is, $f'(x) = 2x$.
    - Iteration steps:
        1. Initialization: Start with an initial guess, $x_{current} = 2$.
        2. Calculate gradient: Calculate the gradient at the current $x$: $x_{current} = 2 * 2 = 4$.
        3. Update $x$: Update $x$ using the formula: $x_{new} = x_{current} - \eta \times \text{gradient}$.
            - $\eta$ = learning rate, a small positive value (e.g., 0.01).
            - $x_{new} = 2 - (0.01 \times 4) = 1.96$.
        4. Repeat: Repeat steps 2 and 3 until x converges to 0, the $x_{new}$ at the end of step 3 will be $x_{current}$ in the next iteration.
    - Convergence:
        - Repeating these steps will cause $x$ to approach 0, which is the minimum of the function.
3. General formula - $x_n = x_c - \eta (\text{slope at x})$:
    - Formula breakdown:
        - $x_n$ = New value of x.
        - $x_c$ = Current value of x.
        - $\eta$ = Learning rate.
        - $(\text{slope at x})$ = Derivative of the function at the current x.
    - Learning rate ($\eta$):
        - The learning rate is a crucial parameter.
        - A small learning rate leads to slow convergence but can prevent overshooting.
        - A large learning rate leads to faster convergence but may cause the algorithm to oscillate or diverge.
    - Application to weights and bias:
        - The same principle applies to optimizing the weights ($w$) and bias ($w_0$) of a model.
        - The partial derivatives of the loss function with respect to $w$ and $w_0$ are calculated and are iteratively updated using gradient descent.
4. Key concept and importance:
    - Optimization: Gradient descent is a fundamental optimization algorithm used in machine learning and deep learning.
    - Iterative process: It's an iterative process that gradually improves the model's parameters.
    - Local minima: Gradient descent can get stuck in local minima, which are not the global minimum.
    - Learning rate tuning: Choosing an appropriate learning rate is essential for successful convergence.
    - Generalization: Gradient descent is a general-purpose algorithm that can be applied to a wide range of optimization problems.

The update rule in the case of loss function:
- $\bar{w} = \bar{w} - \eta(\nabla\bar{w}L(\bar{w}, w_0))$.
- $w_0 = w_0 - \eta(\nabla w_0L(\bar{w}, w_0))$.

# Key Challenges Of Gradient Descent
1. Stopping criteria and iteration count:
    - Forced termination:
        - Gradient descent is an iterative algorithm, meaning it continues to update parameters until a stopping criteria is met.
        - In practice, maximum number of iterations are set to prevent the algorithm from running indefinitely. This "forced stopping" is necessary because convergence is not always guaranteed in a finite number of steps.
    - Overshooting:
        - If the learning rate or number of iterations is too high, the algorithm might overshoot the minimum, leading to oscillations or divergence.
        - Predefining the number of iteration helps to mitigate this risk.
2. Hyperparameter tuning:
    - Importance of tuning:
        - The learning rate (step size) and the number of iterations are crucial hyperparameters that significantly affect the algorithm's performance.
        - "Tried and tested" values might work for some problems, but they are unlikely to be optimal for all.
    - Hyperparameter tuning:
        - Hyperparameter tuning involves systematically searching for the best combination of hyperparameters.
        - Techniques like grid search, random search, and Bayesian optimization are used for this purpose.
        - Finding the correct learning rate is very important. Too high of a learning rate, and the algorithm will not converge, to low of a learning rate, and it will take too long to converge.
3. Initialization:
    - The initial values of the model's parameters can significantly affect the convergence and final performance of gradient descent.
    - Poor initialization can lead to slow convergence, getting stuck in local minima, or even divergence.
4. Local minima and saddle points:
    - Local minima:
        - Gradient descent can get stuck in local minima, which are points where the function's value is lower than its neighboring points, but not the global minimum.
        - This is a particular problem in non-convex optimization problems.
    - Saddle points:
        - Saddle points are points where the function is flat in some directions and steep in others.
        - Gradient descent can get stuck in saddle points, especially in high-dimensional spaces.
5. Sensitivity to data scaling:
    - Gradient descent can be sensitive to the scaling of the input features.
    - Features with larger scales can dominate the optimization process.
    - Feature scaling techniques like standardization and normalization are used to address this issue.

The objective is to find the value of $\bar{w}$ and $w_0$ where the function is minimum.

Use gradient descent to find this. Gradient descent is the optimization algorithm that can help to find the minimum of any given function.

# Objective
To find the gradient of the loss function, i.e.,

$\bar{w} = \bar{w} - \eta(\nabla_{\bar{w}}L(\bar{w}, w_0))$

# Optimization Problem
1. Original optimization problem:
    - Loss function:
        - $\min (- \sum^n_{i = 1} (\frac{\bar{w}^T.\bar{x} + w_0}{||\bar{w}||}) * y_i)$.
        - This function represents the loss that is to be minimized.
        - It involves the weight vector $\bar{w}$, the bias $w_0$, data points $\bar{x_i}$, and labels $y_i$.
        - The term $||\bar{w}||$ (the magnitude or norm of $\bar{w}$) is in the denominator.
    - Minimization goal:
        - The goal is to find the values of $\bar{w}$ and $w_0$ that minimizes this loss.
2. The challenge of $||w||$:
    - Complexity:
        - The term $||\bar{w}|| = \sqrt{w_1^2 + w_2^2 + ...}$ introduces complexity into the loss function.
        - Calculating the derivative (gradient) of the loss function with respect to $\bar{w}$ becomes more challenging due to the square root and the sum of squared components.
        - This increases the computational burden during optimization.
    - Impact on minimization:
        - Minimizing $||\bar{w}||$ contributes to minimizing the overall loss function.
        - A smaller $||\bar{w}||$ makes the denominator smaller, which in turn makes the overall term larger (in absolute value), and since there is a negative sign in front of the sum, it makes the overall sum smaller.
3. Normalizing $||\bar{w}||$ to 1:
    - Simplification:
        - To simplify the optimization problem, the magnitude of $\bar{w}$ can be constrained to 1: $||\bar{w}|| = 1$.
        - This is called as normalization.
        - This removes the square root term from the denominator, making the gradient calculation much simpler.
    - Direction preservation:
        - Setting $||\bar{w}||$ to 1 does not change the direction of $\bar{w}$, it only scales its magnitude.
        - The direction of $\bar{w}$ is what determines the orientation of the hyperplane.
    - Equivalent hyperplane:
        - A hyperplane defined by $\bar{w}$ and $w_0$ is the same hyperplane defined by $\frac{\bar{w}}{||\bar{w}||}$ and $\frac{w_0}{||\bar{w}||}$.
        - Therefore, normalizing $\bar{w}$ does not change the decision boundary.
4. The new optimization problem:
    - Constrained optimization:
        - $\argmin_{(\bar{w}, w_0)_{\text{subject to} ||\bar{w}|| = 1}} - \sum^n_{i = 1} (\frac{\bar{w}^T.\bar{x} + w_0}{||w||}) * y_i$.
        - This is not a constrained optimization problem, where the loss is minimized subject to the constraint $||\bar{w}|| = 1$.
    - Benefits:
        - Simplifies gradient calculations.
        - Reduces computational cost.
        - Does not alter the decision boundary,

# Lagrange Multiplier And Lagrange's Findings
1. Intuitive example - Maximizing $f(x)$ with constraint $g(x)$:
    - Objective function:
        - $f(x) = \frac{4}{x}$ (to be maximized).
    - Constraint function:
        - $g(x) = 8 - 4x$ (must be satisfied).
    - Geometric interpretation:
        - Plot of $f(x)$ looks as follows,

        <img src = "../artifacts/linear_algebra_and_optimization_68.png" width = "400" height = "300">

        - The plot of $f(x)$ and $g(x)$ looks as follows,

        <img src = "../artifacts/linear_algebra_and_optimization_69.png" width = "400" height = "300">

        - The graphs illustrate how the constraint limits the possible values of x.
        - The maximum of $f(x)$ within the constraint occurs at the intersection point of the 2 graphs.
        - This intersection point represents the solution that satisfies both the objective and the constraint.
    - The point of intersection:
        - The whole point of the lagrange multiplier is to find this point of intersection. Without the constraint, the function $f(x)$ would continue to increase as $x$ approaches zero. The constraint limits the search.
2. Tangent slopes and gradients:
    - Equal slopes:
        - At the point of intersection, the slopes of the tangent lines to $f(x)$ and $g(x)$ are proportional.
        - This means that the gradients (which are generalization of slopes to higher dimensions) are also proportional.
    - Proportionality:
        - $\nabla f(x) = \lambda \nabla g(x)$.
        - Where,
            - $\nabla f(x)$ is the gradient of the objective function.
            - $\nabla g(x)$ is the gradient of the constraint function.
            - $\lambda$ is the Lagrange multiplier, a proportionality constant.
    - Lagrange Multiplier's role:
        - $\lambda$ scales the gradient of the constraint to match the gradient of the objective at the optimal point.
        - It essentially tells how much the constraint affects the optimal solution.
3. Mathematical derivation:
    - Derivatives:
        - $f'(x) = - \frac{4}{x^2}$.
        - $g'(x) = -4$.
    - Proportionality equation:
        - $- \frac{4}{x^2} = \lambda (-4)$.
        - $\lambda = \frac{1}{x^2}$.
    - Explanation:
        - This shows how $\lambda$ relates to $x$ at the optimal point.
        - The value of lambda is not always 1. It is a value that scales the gradient of the constraint function.
4. The Lagrangian Function:
    - Combining objective and constraint:
        - $L(\lambda, x) = f(x) + \lambda g(x)$.
        - This Lagrangian function combines the objective and constraint into a single equation.
    - Why combine?
        - Optimization algorithms typically work with a single objective function.
        - The Lagrangian allow to incorporate the constraint into the optimization process.
    - Optimization of Lagrangian:
        - To find the optimal solution, the critical points of the Lagrangian are found by detting its partial derivatives with respect to $x$ and $\lambda$ to zero.
        - This efficiency solves the constrained optimization problem.
5. Lagrange's findings:
    - Finding 1 - Proportional gradients:
        - $f(x) = \lambda g(x)$.
        - This expresses the condition for optimality at the point of intersection.
    - Finding 2 - Lagrangian function:
        - $L(\lambda, x) = f(x) + \lambda g(x)$.
        - This combines the objective and constraint into a single function.

# Applying Lagrange's Findings On The Objective
1. Setting up the Lagrangian:
    - Objective function:
        - $\min L(\bar{w}, w_0) = - \sum^n_{i = 1} (\bar{w}^T.x_i + w_0) * y_i$.
    - Constraint:
        - $||w|| = 1$.
    - Lagrangian:
        - $L(\bar{w}, w_0, \lambda) = - \sum^n_{i = 1} (\bar{w}^T.x_i + w_0) * y_i + \lambda(||\bar{w}|| - 1)$.
        - The constraint is used to form the Lagrangian.
        - Lagrangian can be re-written as,
            - $L(\bar{w}, w_0, \lambda) = - \sum^n_{i = 1} (\bar{w}^T.x_i + w_0) * y_i + \lambda(\sqrt{\bar{w}^T . \bar{w}} - 1)$.
2. Gradient with respect to $\bar{w}$ ($\nabla_{\bar{w}} L$):
    - Partial derivative:
        - $\frac{∂}{∂\bar{w}} L(\bar{w}, w_0, \lambda) = - \sum^n_{i = 1} \frac{∂}{∂\bar{w}}(\bar{w}^T.x_i + w_0) * y_i + \lambda \frac{∂}{∂\bar{w}}(\sqrt{\bar{w}^T . \bar{w}} - 1)$.
    - First term:
        - $\frac{∂}{∂\bar{w}} (\bar{w}^T . x_i + w_0) . y_i = x_i . y_i$.
        - This is because $\bar{w}^T . x_i$ is a dot product, and its derivative with respect to $\bar{w}$ is simply $x_i$.
    - Second term:
        - $\frac{∂}{∂\bar{w}}(\lambda .\sqrt{\bar{w}^T . \bar{w}} - \lambda ) = \frac{\lambda . \bar{w}}{\sqrt{\bar{w}^T . \bar{w}}} = \lambda \frac{\bar{w}}{||\bar{w}||}$.
        - This is the derivative of the norm of $\bar{w}$.
    - Combined gradient:
        - $\nabla_{\bar{w}} L(\bar{w}, w_0, \lambda) = - \sum^n_{i = 1} x_i . y_i + \lambda \frac{\bar{w}}{||\bar{w}||}$.
        - Since $||w|| = 1$,
            - $\nabla_{\bar{w}} L(\bar{w}, w_0, \lambda) = - \sum^n_{i = 1} x_i . y_i + \lambda . \bar{w}$.
3. Gradient with respect to $w_0$ ($\nabla_{w_0} L$):
    - Partial derivative:
        - $\frac{∂}{∂\bar{w}} L(\bar{w}, w_0, \lambda) = - \sum^n_{i = 1} \frac{∂}{∂\bar{w}}(\bar{w}^T.x_i + w_0) * y_i + \lambda \frac{∂}{∂\bar{w}}(\sqrt{\bar{w}^T . \bar{w}} - 1)$.
    - Simplification:
        - Since $\bar{w}^T . \bar{w}$ does not contain $w_0$, the second term becomes zero.
        - $\frac{∂}{∂w_0} L(\bar{w}, w_0, \lambda) = - \sum^n_{i = 1} \frac{∂}{∂w_0}w_0.y_i = - \sum^n_{i = 1} y_i$.
    - Combined gradient:
        - $\nabla_{w_0} L(\bar{w}, w_0, \lambda) = - \sum^n_{i = 1} y_i$.
4. Gradient descent updates:
    - Update rule for $w$:
        - $\bar{w_{new}} = \bar{w_{old}} - \eta(\nabla_{\bar{w}} L(\bar{w}, w_0, \lambda))$.
        - $\bar{w_{new}} = \bar{w_{old}} - \eta(- \sum^n_{i = 1} x_i . y_i + \lambda . \bar{w_{old}})$.
    - Update rule for $w_0$:
        - $w_{0_{new}} = w_{0_{old}} - \eta(\nabla_{w_0} L(\bar{w}, w_0, \lambda))$.
        - $w_{0_{new}} = w_{0_{old}} - \eta(- \sum^n_{i = 1} y_i)$.
5. Lagrange Multiplier ($\lambda$):
    - The Lagrange Multiplier must be solved for in order to have a complete solution.
    - The solution for $\lambda$ is found by using the constraint that will $||w|| = 1$.
    - Once $\lambda$ is found, then the $\bar{w}$ and $w_0$ values can be correctly calculated.

# Types Of Gradient
1. Batch Gradient Descent (BGD):
    - Concept:
        - BGD calculates the gradient of the cost function with respect to the parameters using the entire training dataset in each iteration.
        - It then updates the parameters based on this gradient.
    - Pros:
        - Guaranteed convergence to the global minimum for convex cost functions.
        - Stable convergence.
    - Cons:
        - Very slow for large datasets, as it requires processing the entire dataset in each iteration.
        - Computationally expensive.
        - Cannot be used for online learning (where data arrives sequentially).
    - Use cases:
        - Suitable for small datasets where computational cost is not a major concern.
2. Stochastic Gradient Descent (SGD):
    - Concept:
        - SGD calculates the gradient using only a single randomly selected data point from the training set in each iteration.
        - It updates the parameters based on this single-point gradient.
    - Pros:
        - Much faster than BGD, especially for large datasets.
        - Can be used for online learning.
        - Can escape local minima more easily due to its noisy updates.
    - Cons:
        - Noisy updates can lead to oscillations and slower convergence.
        - May not converge to the exact minimum but rather oscillate around it.
    - Use cases:
        - Suitable for large datasets and online learning applications.
3. Mini-Batch Gradient Descent (MBGD):
    - Concept:
        - MBGD is a compromise between BGD and SGD.
        - It calculates the gradient using a small batch of randomly selected data points (a mini-batch) in each iteration.
        - It updates the parameters based on the mini-batch gradient.
    - Pros:
        - Faster than BGD and more stable than SGD.
        - Balances the advantages of both BGD and SGD.
        - Efficient matrix computations can be performed on mini-batches.
    - Cons:
        - Requires tuning the mini-batch size hyperparameter.
        - Still has some noise in the gradient.
    - Use cases:
        - The most commonly used type of gradient descent in deep learning.
