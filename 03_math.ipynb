# Number System
A number system is a way of representing numbers using digits or symbols, following a consistent set of rules. It defines how numbers are written and what each digit's value means based on its position within the number and the base of the system.

- Decimal numbers: $(29)_{10}$, $(1996)_{10}$, ...
- Binary numbers: $(11101)_{2}$, $(1111)_{2}$, ...

# Ranges
- Closed interval - $[1, 10]$: $\text{Range} = \text{End} + \text{Start} - 1$.
- Open interval - $(1, 10)$: $\text{Range} = \text{End} - \text{Start} - 1$.

# Logarithms
A logarithm is a mathematical way of expressing exponents, or the power to which a number must be raised to get another number. Logarithms are the inverse of exponents.

In the equation below, $b$ is the number of times $a$ has to be multiplied with itself for in order to get $x$,
- $\log_{a} x = b$.

The above is equivalent to,
- $e^y = x$.

# Combinatorics
Combinatorics is the art of counting.

### Permutations
- Permutations deals with the arrangement of objects.
- In permutations, the order of arrangement matters, $i, j ≠ j, i$.
- $n_{P_r} = \frac{n!}{(n - r)!}$. Where,
    - $n$ = Number of objects.
    - $r$ = Number of slots.
- Total number of ways of arranging $n$ objects around a circle = $(n - 1)!$.

### Combinations
- Combinations deals with the selection of objects.
- In combinations, the order of arrangement does not matter, $i, j = j, i$.
- $n_{C_r} = \frac{n!}{(n - r)!r!}$. Where,
    - $n$ = Number of objects.
    - $r$ = Number of objects to be selected.
- Properties:
    - $n_{C_0} = 1$.
    - $n_{C_1} = n$.
    - $n_{C_0} + n_{C_1} + ... + n_{C_n} = 2^n$.

### Circular Permutations
In circular permutations, the arrangement of objects around a circular structure matters, but rotations are considered the same arrangement. For n distinct objects, the number of circular permutations is $(n - 1)!$, if clockwise and counterclockwise arrangements are distinct.

# Calculus
- Differential calculus makes use of derivatives to measure the instantenous rate of change, like finding the slope of a tangent to a curve.
- Integral calculus makes use of integrals to calculate the accumulation of quantities, like area under the curve.
- A function is a mapping between inputs and output.
- Types of functions:
    1. Polynomial: A polynomial function is a mathematical function formed by adding terms that consists of a coefficient multiplied by a variable raised to a non-negative integer power. For example, $f(x) = 4x^{2} - 3x + 2$.
    2. Monomial: A monomial function is a polynomial function with only 1 term, consisting of a constant, a variable, or a product of a constant and a variable raised to a positive integer power. For example, $f(x) = 4x^{3}$.
    3. Trigonometric: Trigonometric functions, also know as circular functions, are functions that relate the angles of a right-angled triangle to the ratios of its sides, including sine, cosine, tangent, cotangent, secant, and cosecant.
        - $\sin(x)$ The sine function is defined as the ratio of the length of the opposite side to that of the hypotenuse in a right-angled triangle. The sine function is used to find the unknown angle or sides of a right triangle.
            - $\sin\theta = \frac{Opposite}{Hypotenuse}$
            - $\text{Domain} = (-\infty, \infty)$
            - $\text{Range} = [-1, 1]$
        - $\cos(x)$: The cos function is defined as the ratio of the length of the adjacent side to that of the longest side, i.e., hypotenuse.
            - $\cos\theta = \frac{Adjacent}{Hypotenuse}$
            - $\text{Domain} = (-\infty, \infty)$
            - $\text{Range} = [-1, 1]$
        - $\tan(x)$: The tan of an angle is the ratio of the length of the opposite side to the length of the adjacent side.
            - $\tan\theta = \frac{Opposite}{Adjacent}$
            - $\tan\theta = \frac{sin\theta}{cos\theta}$
            - $\text{Domain} = (-\infty, \infty)$
            - $\text{Range} = (-\infty,\infty)$.
    4. Exponential: Exponents, also known as powers, are a shorthand way to represent multiplication of the same number by itself, where the number being multiplied is the base and the exponent indicates how many times to multiply the base by itself. Exponential functions are of the following form, $f(x) = e^{x}$.
    5. Logarithmic: A logarithm is a mathematical way of expressing exponents, or the power to which a number must be raised to get another number. Logarithms are the inverse of exponents. Logarithmic functions have the following form, $f(x) = ln(x)$.
- Domain is the range of values that a function can accept as input.
- Range is the range of values that a function returns as output.
- A function is continuous if, $LHL = RHL = f(x)$.
- For a function to be differentiable, it has to be continuous.
- Ab-initio method is used to find the derivative of any function under the sun. It is defined as,
    - $f'(x) = \lim\limits_{h \to 0}\frac{f(x + h) - f(x)}{h}$.
- Properties of derivatives:
    - Linearity: If $h(x) = f(x) + g(x)$, then $h'(x) = f'(x) + g'(x)$.
    - Product rule: If $h(x) = f(x) * g(x)$, then $h'(x) = f'(x)g(x) + f(x)g'(x)$.
    - Quotient rule: If $h(x) = \frac{f(x)}{g(x)}$, then $h'(x) = \frac{f'(x)g(x) - f(x).g'(x)}{g(x)^2}$.
    - Chain rule: If $h(x) = f(g(x))$, then $h'(x) = f'(g(x)) * g'(x)$.

# Probability
- Probability is a value between 0 and 1 which defines the likelihood of an event occurring. 0 represents a highly unlikely event, and 1 represents a highly likely event.
- Sample space is a set containing all the possible outcomes of an experiment.
- A experiment is a procedure that can be repeated for infinite number of times, and has a well defined sample space.
- An event is a subset of a sample space.
- Set operations:
    - Union.
    - Intersection.
    - Compliment.
    - Symmetric difference.
- $P(A \cup B) = P(A) + P(B) - P(A \cap B)$.
- $P(A \cap B) = P(A) + P(B) - P(A \cup B)$.
- A collection of sets are disjoint, if the intersection of all of the sets in the collection is an empty set or a null set. An empty set or null set is denoted by $\phi$.
- A collection of sets are pairwise disjoint, if the intersection of any 2 sets in the collection is an empty set.
- A collection of sets that are pairwise disjoint, are also simply disjoint because the intersection of all of the sets results in an empty set.
- When the union of 2 or more sets results in the universal set ($U$), and no element is common to any pair of the sets, they are said to be mutually exclusive and collectively exhaustive sets.
- Conditional probability measures the probability of occurrence of an event given that another event has already occurred. It is mathematically expressed as,
    - $P(A | B) = \frac{P(A \cap B)}{P(B)}$ (Used when the probabilities are given).
    - $P(A | B) = \frac{N(A \cap B)}{N(B)}$ (Used when the frequencies are given).
- Law of total probability calculates the collective sum of probabilities of an event in each partition of the universal set.

    

    - $P(A) = \sum P(A \cap B_i)$.
    - $P(A) = \sum P(A | B_i) \times P(B_i)$ (simplified from conditional probability's expression).
- Bayes theorem defines $P(B | A)$, if $P(A | B)$ is known.
    - $P(B | A) = \frac{P(A | B) \times P(B)}{P(A)}$. Where,
        - $P(A|B)$ = Posterior. The probability of A being true, given B is true.
        - $P(B|A)$ = Likelihood. The probability of B being true, given A is true.
        - $P(A)$ = Prior. The probability of A being true. This is knowledge.
        - $P(B)$ = Marginalization. The probability of B being true.
- 2 events $A$ and $B$ are independent if $A$ does not convey any information about $B$.
    - $P(A | B) = P(A)$, and $P(B | A) = P(B)$.
    - $P(A \cap B) = P(A) \times P(B)$.

# Statistics
- Descriptive statistics define informational coefficients that help in summarizing a dataset. These coefficients may be representative of the entire population or a sample of population.
- Descriptive statistics are broken down into measures of central tendency or measures of variability.
- Inferential statistics uses measurements from the samples of subjects from the experiment to compare treatment groups and make generalizations about the larger population of subjects.
- Mean is the average of all data points in a dataset.
- Median is the data point that occurs at the center of a sorted dataset.
- Mode is the most recurring value in a dataset.
- Percentage is used to represent the fraction of space occupied by an entity out of the total space.
- Percentile defines how much better an entity is in comparison to others.
- Standard deviation defines the amount of spread in the data. It measure the dispersion of data points in a dataset relative to the mean and is calculated as the square root of the variance.
    - $\sigma = \sqrt{\frac{\sum (x_i - \bar{x})}{(n - 1)}}$.
- Variance also describe the amount of spread in the data.
    - $\sigma^2 = \frac{\sum (x_i - \bar{x})}{(n - 1)}$.
- Standard deviation is expressed in the same units as the original data.
- Variance is expressed in squared units of the original data.
- IQR defines the range or the spread of the middle half of the distribution.
    - $IQR = Q_3 - Q_1$.
- Any data points greater than upper fence and lower than the lower fence are treated as outliers.
    - $\text{Upper Fence} = Q_3 + (IQR \times 1.5)$.
    - $\text{Lower Fence} = Q_1 - (IQR \times 1.5)$.
- 1.5 is a constant used to discern the outliers. The number 1.5, hereafter scale, clearly controls the sensitivity of the range and hence the decision rule. A bigger scale would make the outliers be considered as inliers, while a smaller scale would make some of the data points be perceived as outliers. And, none of these cases is desirable.
- Outliers are handled by removing them from the dataset.
- Skew in distribution is the asymmetry observed in the probability distribution.

    

### Gaussian Distribution
- Gaussian distribution is also called normal dirtribution. The shape of the Gaussian resembles a bell.
- The shape of the Gaussian becomes smoother as the number of bins increase. The bin size decreases as the number of bins increase.
- The highest number of data points occur around the mean in a Gaussian.
- If the standard deviation ($\sigma$) is given as 2.5, then each step taken to the left or right of the mean, represents a change in 2.5 units in the measurement. Meaning, if $\sigma = 65$, then $+1\sigma = 67.5$ and $-1\sigma = 62.5$.
- 68% of the data points lie within the range of $1\sigma$ to the left or right of the mean.
- 95% of the data points lie within the range of $2\sigma$ to the left or right of the mean.
- 99.7% of the data points lie within the range of $3\sigma$ to the left or right of the mean.
- The 68-95-99.7 rule is called as the empirical rule in statistics.
- Z-Score or Standard Score quantifies how many standard deviations does a data point lie away from the mean.
    - $\text{Z-Score} = \frac{X - \mu}{\sigma}$.
- CDF describes the cumulative probability distribution of a random variable. It is used to find the probability that a random variable will take on a value which is less than or equal to the given value. It takes Z-Score as the input and returns a fraction representing the probability as output.
- PPF is the inverse of CDF. It is used to find the Z-Score of a value for which the probability of a random variable is reached or exceeded. It takes the probability value as input and returns Z-Score as output.

### Central Limit Theorem And Bootstrapping
- CLT states that for a sufficient large sample size taken from a population, with finite amount of variations, the mean of all the sampled variables from the same population will be approximately equal to the mean of the population. Furthermore, the distribution of the samples approximates the normal distribution, and the variances of the sample also approximates the variances of the population, as the sample size increases or gets larger. CLT is mathematically defined as,
    - $\bar{X} \sim N(\mu, \frac{\sigma}{\sqrt{n}})$.
- Components of CLT:
    - Sampling is successive.
    - Sampling is random.
    - Sampling is independent.
    - Sampling should be limited.
    - Sampling size is increasing.
- CLT is used when the discussion is about mean.
- Bootstrapping is used when the discussion is about other statistical entities.
- In Statistics and Machine Learning, bootstrapping is a resampling technique that involves repeatedly drawing samples from the source data with replacement, often to estimate a population parameter.
- Sampling is done with replacement in both, i.e., CLT and Bootstrapping.

### Confidence Interval And Significance Level ($\alpha$)
- Confidence Interval is the mean of the estimate +/- the variation in that estimate. Essentially, Confidence Interval defines a range of values within which the estimate of the experiment may fall, if the experiment is re-run.
- Significance Level ($\alpha$) is the threshold that is set for making decision about the null hypothesis. The value set as $\alpha$ represents the probability of making type 1 errors (this represents the probability of H0 being rejected even when it is true).
- The relationship between Confidence Interval and Significance Level is,
    - $CI = 1 - \alpha$.

### Bernoulli Distribution
- Bernoulli distribution is used to model experiments involving 1 trial and 2 outcomes. The 2 outcomes are "success" and "failure".
- The probability of success is $p$.
- The probability of failure is $(1 - p)$.
- Each trial is considered to be independent in Bernoulli's distribution.

### Binomial Distribution
- Binomial distribution describes the probability of obtaining certain number of successes in a sequence of $n$ independent trials, each of which have only 2 outcomes. The 2 outcomes are "success" and "failure".
- The probability of success is $p$.
- The probability of failure is $(1 - p)$.
- The number of trials are fixed.
- Each trial is considered to be independent of the other.
- $p$ is constant for each trial.
- A Probability Mass Function (PMF), also sometimes called as probability function or frequency function, is a mathematical tool used in probability and statistics specifically for discrete random variables. It describes the probability of each possible value that the random variable can take.

### Geometric Distribution
- Geometric distribution describes the probability of number of trials needed to get the first success in a sequence of independent and identically distributed Bernoulli trials.
- Each trial has only 2 outcomes, "success" and "failure".
- Each trial is independent of the other.
- The probability of success remains the same for each trial.

### Poisson Distribution
- Poisson distribution describes the probability of a given number events occurring in a fixed interval of time or space if these events occur with a known average rate and independently of the time since the last event.
- Each event is independent of the other.
- The events happen at random within the given interval.

### Exponential Distribution
- Exponential distribution deals with the amount of time until some specific event occurs. For example, the amount of time a device operates for before failing.
- It is very closely related to Poisson distribution. While Poisson distribution deals with the number of occurrences of an event in a given time interval, exponential distribution deals with the time between those events.
- Exponential distribution has a memoryless property. Meaning, the probability of an event happening in the next moment is independent of how much time has elapsed.
- Rate ($\lambda$) is the average number of times an event will occur per unit time.
- Scale is the amount of time between 2 events.
    - $\text{Scale} = \frac{1}{\lambda}$.

### Log-Normal Distribution
- A Log-Normal distribution is a continuous probability distribution of a random variable whose logarithm is normally distributed.
- If the natural logarithm of each data point is taken in a Log-Normal distributed dataset, the resulting transformed data will follow a normal distribution.
- The Log-Normal distribution is typically right skewed.

### CDF vs. PPF vs. PMF
- CDF helps in calculating the probability that random variable will take on a value that is less than or equal to a certain value.
- PPF helps in calculating the Z-Score of the value below which a certain fraction of values lie.
- PMF helps in calculating the probabilities of each possible value that a random variable will take.
