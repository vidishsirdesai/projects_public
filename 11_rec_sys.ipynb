# Market Basket Analysis
Consider departmental stores like Big Bazaar, SPAR, Reliance Fresh, Walmart, 7Eleven, etc. The layout of these stores is designed in a certain fashion. For example, the billing counter is placed in a certain location within the store, the various dedicated shelves for different items (confectionary, milk, fruits, vegetables, etc) are placed in a certain location.

The placement of these items in the store is based on the likelihood of them occurring together. The question, "What other items will a customer purchase, if they have purchased an item?", is what defines the layout or the design of these stores.

Now consider, Google search as example, when a user clicks on the search bar, a drop down list with recommendations on what the trending searches are is shown. These recommendations listed, are user specific.

Such systems, which hand out recommendations to the users by analyzing their interactions or transactions are essentially what Recommender Systems are. Market Basket Analysis is the simplest kind of Recommender System.

Consider a grocery store, and consider 1 user, for example. The following are the transactions they made,

| dates | items |
| :-: | :-: |
| 01-03-2024 | Bread, Milk, Eggs, Butter |
| 03-03-2024 | Milk, Bread, Grapes, Yogurt |
| 05-03-2024 | Milk, Bread, Buttermilk, Lemon |
| 05-03-2024 | Peanuts, Onion, Pringles |

Where, 
- $\text{1 transaction} = \text{the customer enters store} + \text{the customer searched for an item and added it to the cart} + \text{customer made payment} + \text{customer exited the store}$.

The following is how Market Basket Analysis works,
- The past purchase behavior of the customer is considered, and analyzed across all the transactions, and the things that customer, or customers are buying together are understood. For example, in the analysis, if it is found that if milk, and bread are bought together, then it makes sense to place milk, and bread in the same shelf, or a shelf close to the other.
- In a different scenario, if it is found that toothpaste, and toothbrush are bought together, then it makes sense to place them both close by. In addition to toothpaste, and toothbrush, all the grooming products can be placed close by.

This above, is a high level understanding of what Market Basket Analysis helps in achieving.

# Stock Keeping Unit (SKU)
Stock Keeping Unit gives information about the kind of commodities that are in stock. Meaning, if there are Digestive Biscuits, then that is 1 SKU, or if there is Breakfast Tea, then that is 1 SKU. During billing, the item is scanned under a bar code reader to fetch the price of that item.

SKU in simple terms terms gives information about weather an item is in stock or not. Each item is 1 SKU. In a local store there could be some 10 to 20 thousand SKUs. In a mega store like, Amazon, there could be a million SKUs. In 1 transaction, a user could buy 10 - 20 SKUs, or distinct products.

Consider that there are $m$ transactions made in a month,
- $T_1 = {a_1, a_2, a_5}$,
- $T_2 = {a_3, a_7, a_n}$,
- ...
- $T_i = {a_4, a_8, a_2, a_5, a_i}$
- Where,
    - $T_i$ = Transaction.
    - $a_i$ = SKU.

Say that there are $D$ items in total, which is the total number of SKUs. Meaning, if there are $n$ distinct products, it means $D = {1, 2, 3, ..., n}$. Every time a transaction is done it is a subset of $D$. Meaning, $T_i$ is a subset of $D$.

Consider an example where there are $n$ items, $T_1 = {a_1, a_2, a_3}$. $T_1$ is a subset of D.

Once the information of all the transactions are available, the items that people bought together can be grouped (groups are also called item sets). For example,

- Raeesa = {"Milk", "Bread", "Banana", "Honey"}

One observation from the above examples is that, whenever a person bought Milk, they also bought Bread. This observation, is an outcome of Market Basket Analysis.

According to Market Basket Analysis, each transaction is basket, i.e., $T_1$, $T_2$, $T_3$, ..., $T_i$, and if they are analyzed, and patterns of what items are bought together are observed, then directions can be given to the superstore on how the items should be placed. Together, or in close proximity to each other.

# Data Transformation In Market Basket Analysis
- Whenever Market Basket Analysis is done, quantity is not considered. Meaning, if a person is buying 6 Beers and 1 Diaper, and if the other person is buying 1 Beer and 2 Diapers, both the transactions are considered equal. This is because, the objective is to find the items which are purchased together. The quantity of items sold is therefore not important.
- Market Basket Analysis for each store, or each region is performed separately.
- The returned items should be dropped from the analysis.
- A Sparse Martix, where the rows represent the `InvoiceNumber`, and the columns represent the `SKUs`, is created.
- The Sparse Martix is populated in the following fashion,
    - For each `InvoiceNumber`, the `SKUs` that were purchased are marked as `1`, and the remaining `SKUs` are maked as `0`.
- The following is an example of how the Sparse Martix looks,

| InvoiceNo | bread | butter | egg |
| :-: | :-: | :-: | :-: |
| $a_1$ | 1 | 0 | 0 |
| $a_2$ | 1 | 1 | 1 |
| $a_3$ | 0 | 0 | 1 |
| . | | | |
| . | | | |
| . | | | |
| $a_m$ | 1 | 0 | 1 |

- The Sparse Matrix is called so, because majority of the entries are `0`.
- The transactions where only 1 items were purchased are dropped.

# Apriori Algorithm
### Steps
1. Determine the support for each individual item (1-itemsets):
    - Iterate through the entire transaction dataset.
    - Count the number of times each individual item appears in the transactions.
    - Compare the support count of each item with the minimum support threshold (defined by the user).
    - Identify the frequent 1-itemsets (L1) – those items whose support is greater than or equal to the minimum support.
2. Generate candidate itemsets of size k (Ck) from frequent itemsets of size k-1 (Lk-1):
    - This step involves joining the frequent itemsets of size k-1 to create potential candidate itemsets of size k.
    - Two frequent (k-1)-itemsets are joined only if their first k-2 items are the same. For example, if {A, B} and {A, C} are frequent 2-itemsets, they can be joined to form a candidate 3-itemset {A, B, C}.
    - This joining process aims to create all possible candidate k-itemsets.
3. Prune candidate itemsets (Ck):
    - For each candidate k-itemset generated in step 2, check if all of its (k-1)-subsets are present in the frequent (k-1)-itemsets (Lk-1).
    - If any (k-1)-subset of a candidate k-itemset is not frequent, then the candidate k-itemset itself cannot be frequent (based on the Apriori property) and is thus pruned (removed). This step significantly reduces the number of itemsets to be checked in the next step.
4. Calculate the support for each candidate itemset in Ck:
    - Iterate through the entire transaction dataset again.
    - For each transaction, check which of the candidate k-itemsets are present in that transaction.
    - Count the support for each candidate k-itemset.
5. Identify the frequent k-itemsets (Lk):
    - Compare the support count of each candidate k-itemset in Ck with the minimum support threshold.
    - Identify the frequent k-itemsets (Lk) – those candidate itemsets whose support is greater than or equal to the minimum support.
6. Repeat steps 2-5 until no new frequent itemsets can be generated.
    - The algorithm terminates when the set of frequent k-itemsets (Lk) becomes empty, meaning no more larger frequent itemsets can be found.

### Example
Consider that a store is being run where there are 3 items sold. There are 6 total possible combinations in which the items can be sold in,
1. {Milk}
2. {Egg}
3. {Bread}
4. {Milk, Egg}
5. {Milk, Bread}
6. {Bread, Egg}
7. {Milk, Egg, Bread}
8. {}

No consider,
- If there are $n$ number of items in a store, then the total number of combinations that a customer can choose from are $2^n$.
- And if there are $m$ number of transactions per month, the the time complexity of traversing through the whole dataset will be $(2^n) * m$.

Consider that, there are 1 million items, and there are 10 transactions per month. The total possible transactions = $(2^{10^6}) \times (10)$.

This is an astronomically high time complexity. Therefore, a threshold called Support is set or defined. According to Support, if there are $n$ items, then only $k$ items will be considered, $k$ being the set threshold.

Consider Milk and Bread for example. Only items which have been sold for 10% of the times are considered. Meaning, if there are 1000 transactions, and if Milk does not appear in 100 transactions at least, then Milk will not be considered.

Now consider another example where a store is being run with 4 items. The number of possible combinations in which these items can be bought = $2^4 = 16$.

The possible combinations are,
- {1}, {2}, {3}, {4}, {1, 2}, {1, 3}, {1, 4}, {2, 3}, {2, 4}, {3, 4}, {1, 2, 3}, {1, 3, 4}, {2, 3, 4}, {1, 2, 3, 4}, {}.

Say that the threshold is 100. Meaning, if a set is not sold for 100 times, it will not be considered.

Consider hypothetically that there are 500 transactions. The question is, out of theses 500 transactions how many times is 1 sold (1 could be sold alone, or 1 could be sold in a group with other items)? If 1 is sold as {1}, {1, 2}, {1, 2, 3}, then 1 is counted to be sold for 3 times.

If 1 is sold for 110 times, then 1 should be considered. If 4 is sold only for 50 times, then it is not considered for Market Basket Analysis. This also implies that if 4 is sold along with 1, then that combination is not considered as well. The same is the case with all other combinations which have 4 (they are not considered).

Next consider the combination {1, 2}, the number of times 1 and 2 are bought together is checked. If it is below the threshold then it is dropped. The same is repeated for all other combinations as well. Only the combinations which are above the threshold are kept for analysis, and the rest are discarded. This algorithm for Market Basket Analysis is a quicker way. Else, the algorithm will run for $(2^n) * m$ times, which of course requires a lot of computation power. The algorithm used for Market Basket Analysis is called as Apriori algorithm. This is one of the earliest algorithms developed in 1994 - 1995.

# Association Rules
Association rules are an outcome of Apriori algorithm.

Consider a set of all items, $D = {1, 2, ..., n}. Let $x = {1, 2, 3}$, and $y = {4, 6}$.

When there is only 1 item in the item set, there is no association, since it is not paired with anything else. This is contrary to the case where there is a combination like {Bread, Eggs}, and the Support for this is 13%. Meaning, there are 13% cases where Bread and Eggs are bought together. By extension, if a user buys Bread, then there is 13% chance that they will buy Eggs as well.

If Bread is bought 18% of the times, and Bread and Eggs are bought 13% of the times, then the probability that Eggs are bought after Bread is bought is $\frac{13}{18}$. The reverse of this case does not hold. Meaning, it is incorrect to state that if Eggs are bought, then Bread is bought.

To sum up, whenever a pair of item is found, ${a, b}$. It is correct to state that "$b$" is bought, when "$a$" is bought. The reverse of this is not true. Here,
- "$a$" is called Antecedant.
- "$b$" is called Consequent.

# Matrices To Understand The Association
So far, the only matrix seen is called support, but to understand market basket analysis and how thing are related, there are multiple matrices that help in understanding this.

A method called association_rules(), which has a bunch of matrice which help in understanding the relationship between 2 items in a market basket. The matrices are,

### Antecedants
Antecedant is the first item bought in a pair or item support.

### Consequents
Consequent is the second item bought in a pair or item support. If antecedant is bought, then consequent will be bought.

### Antecedant Support
Say that antecedant is represented as X, and consequent as Y.

$\text{Antecedant Support} = \frac{\text{Number of Times X Occurs}}{\text{Total Transactions}}$

### Consequent Support
Say that antecedant is represented as X, and consequent as Y.

$\text{Consequent Support} = \frac{\text{Number of Times Y Occurs}}{\text{Total Transactions}}$

### Support
Say that antecedant is represented as X, and consequent as Y.

$Support = \frac{\text{Number of Times the Pair Occurs}}{\text{Total Transactions}} = \frac{X \cap Y}{N}$

Where,
- $X \cap Y$ = X and Y are bought together.

### Confidence
Conidence explains the probability of Y being bought, if X is bought.

$Confidence = \frac{\text{Number of Times the Pair Occurs}}{\text{Number of Times X Occurs}} = \frac{(X \cap Y)}{\text{Number of Times X Occurs}}$

### Lift
Consider that Milk is a popular product, and Bread is also a popular product. There is a possibility that purchse of Milk is not driving the sale of Bread, because anyways both of them are a popular product. Lift is used to understand such cases.

Consider that there are Green and Pink, it is observed that in 61% of cases, people buy Pink once they buy Green. If people have not bought Green, then how many times would they buy Pink? Independent probability is calculated to understand,
- $\text{Lift} = \frac{X \cap Y}{X * Y}$.

If Y is a very popular item, then the value of Y will be high. Then $(X \cap Y)$ << Y. However, if Y is only popular due to X, then $(X \cap Y)$ ≈ Y.

$P(X \cap Y)$ is high, then every time X is bought, then there is a very high chance of Y being purchased. Then there is also a possibility that Y is also very popular. In general the probability of Y being bought is very high. Then whatever correlation that is seen is due to random chance.

Hence a new metric called lift is devised,
- $Lift = \frac{P(X \cap Y)}{P(X) * P(Y)}$.

If Y is very popular, the $P(X \cap Y)$ << P(Y). However, if Y is only popular with X, then $P(X \cap Y)$ ≈ P(Y).

Where,
- $P(X \cap Y)$ = Support of X and Y.

Therefore,
- $\text{Lift} = \frac{Support(X \cap Y)}{Support(X) * Support(Y)}$.

- If Lift > 1, it is indicative of a positive correlation.
- If Lift = 1, it is indicative of no correlation.
- If Lift < 1, it is indicative of a negative correlation.

### Leverage
Leverage, like Lift, is also constructed using Support. This is very similar to Lift.

The Leverage of "if X then Y", i.e., $(X -> Y)$ is calculated as,

$\text{Leverage}(X -> Y) = \text{Support}(X \cap Y) - (\text{Support}(X) * \text{Support}(Y))$

What is the advantage of using Leverage over Lift values?
- Leverage is easier to interpret.
- Leverage value lies in the range of -1 to +1, where as, Lift value ranges from 0 to infinity.

### Conviction
Conviction is calculated as the ratio of expected frequency that X occurs without Y, if X and Y were independent divided by the observed frequency of incorrect prediction.

$\text{Conviction}(X -> Y) = \frac{(1 - S(Y))}{(1 - C(X -> Y))}$

NOTE: A high value means that the consequent depends strongly on the Antecedant.
