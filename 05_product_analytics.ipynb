# Product Analytics
- Metric is a quantitative measure of an entity.
- Approach to solving product metric questions:
    - Ask clarifying questions about the product or the feature (use the 5Ws and 1H framework).
    - Plan the metric based on the answers.
    - Conclude or summarize the key points.
- How to measure the product's performance?
    1. Clarify about the product or feature (like target audience, need, etc).
    2. Understand the kinds of groups that benefit from the product (like advertisers, users). Also understand the kind of benefits each group has from the product.
    3. Business goals or simply goals that each group using the product intends to achieve.
    4. Finally define the metrics.
- While determining product metrics, a framework or a funnel with 6 stages of metrics is adopted called AAARRR Funnel or the Pirate's Funnel.
- AAARRR or Pirate's funnel is used to find the weakest points in a business that need attention.
- The 6 metrics of the Pirate's Funnel are:
    1. Awareness: How many people are aware of the product.
    2. Acquisition: How many people have visited the website.
    3. Activation: How many people have taken the first step (like app installation, sign up, post comment, etc).
    4. Retention: How many people have come back for the second, third or more times?
    5. Revenue: How many people have started paying and how much do they pay?
    6. Referral: How many people have referred the product to people in their network?
- Success metrics:
    - These are the primary metrics that define the desired outcome of a change or experiment. They directly measure whether the intended goal has been achieved or not.
    - They are key indicators that determine if a product is considered "successful" or not.
- Guardrail metrics:
    - These are the metrics that should be monitored to ensure that a change does not negatively impact other critical areas of the business.
    - They act as "safety checks" to prevent unintended consequences.
- Churn or Churn Rate is the measure of the proportion of population or item moving out of a group over a specific period of time.
    - $\text{Churn} = \frac{\text{Total customers at the end of the time period}}{\text{Total customers at the start of the time period}} \times 100$.
- Product metrics are quantifiable data points that a business tracks and analyzes to evaluate the success of its products. Monitoring and analysis of these metrics helps in making intelligent decisions throughout the product development process.
- KPIs give quantifiable evidence about which aspects of the product or customer experience are resonating with the customers and which aspects are not.
- Classification of product metrics can be based on:
    - Goal.
    - Task.
    - Type of business.
- Levels of product metrics:
    - North star metric:
        - North star metric or focus metric for an organization is the single most important measure of success that matters a lot to the organization. Examples:
            - Instagram/ Facebook: DAU, WAU, MAU.
            - Spotify: Time spent listening.
            - AirBnB: Number of booked nights.
            - Quora: Number of questions answered.
    - Level 1 metric:
        - This metric depicts the desired outcome of a particular product, team or initiative.
        - Level 1 metric should be chosen such that it contributes towards measuring the success of the north star metric.
        - For example, if the product's north star metric is WAU, then a good level 1 metric would be to track 7 day retention to ensure that the marketing funds are not being spent towards acquiring new customers who leave after 2 to 3 days.
    - Level 2 metric:
        - Level 2 metrics are indicators that level 1 metrics are moving in the right direction.
        - These metrics are set to track initiatives at the most granular level.
        - For a product whose north star metric is WAU, and level 1 metric is 7 day retention, then a good level 2 metric would be to track iOS or Android users retention.
- Tooltip explains what a feature or product is and how it should be used. It is launched each time a new product, feature or a version of the same product or feature is revealed to the users.
- Framework for working with problems of product metrics:
    1. Clearly define everything about the feature.
    2. Determine the goal of the feature.
    3. Define every aspect of the customer's journey from end-to-end.
    4. Map and quantify the user behavior at each stage in the user journey.
    5. Evaluate the metrics.
- Retention is when people or users continue to use the application on a regular basis, and return to the app frequently. Retention rate is mathematically expressed as,
    - $\text{Retention Rate} = \frac{\text{Number of users who continued to use the product after the first time}}{\text{Total number of first time users}} \times 100$.
- Resurrection is when a user stops using the product for some time and decides to start using it again after a while.
- Cohort analysis is used to track the behavior of a group of users overtime.
- Funnel analysis is used to track the progression of users through a predefined series of events or steps in specific order.
- Root cause is the underlysing or the core reason for a problem to exist.
- Root Cause Analysis (RCA):
    - RCA is the systematic process of identifying the underlying causes of a problem or an issue.
    - It is defined as a collective term that describes a wide range of approaches, tools and techniques used to uncover the causes of a problem.
    - RCA involves breaking down the problem into smaller parts, examining each part in detail and identifying the root causes that are contributing to the problem.
- Goal of RCA:
    1. The first goal of RCA is to discover the root cause of the problem or an event.
    2. The second goal is to fully understand how to fix, compensate or learn from any underlying issues within the root cause.
    3. The third goal is to apply whatever has been learnt from the analysis to systematically prevent future issues or to repeat success.
- Lifecycle of RCA (CRIEd framework):
    1. Clarify: Ask clarifying questions about the problem and share the thoughts about it.
    2. Rule out: Rule out any change in the metric happening due to technical issues or infrastructural glitches or bugs or outliers.
    3. Internal factors: Explore the internal factors that could have affected the metric. Use the TROPiCS framework:
        - Time.
        - Region.
        - Other related features affected.
        - Platform.
        - Cannibalization.
        - Segmentation.
    4. External factors: Understand the external factors like PR, competitors, seasonality.
- Customer Relationship Management (CRM):
    - CRM combines practices, strategies and technologies that companies use to manage and analyze customer interactions and data throughout the customer's lifecycle.
    - CRM helps the business gain insights into the behavior of their customers and modify their business operations to ensure that customers are serviced in the best possible ways.
- How is CRM used in business?
    - CRM analytics gives insights into customers and how well the sales and customer service teams are reaching them.
    - CRM helps to analyze customer's habits to re-target and optimize sales of existing customers.
    - CRM can be used to find customers with similar behaviors which helps in organizing campaings suitable for each group of customers.
    - CRM helps in improving profitability, productivity and efficiency.
- Customer segmentation involves segregating customers into certain groups based on similarities they share. The segregation can be based on:
    1. Demographics.
    2. Psycographics.
    3. Geography.
    4. Purchase patterns.
    5. Customer preferences.
- RFM Clustering:
    - RFM stands for Recency, Frequency and Monetary Value.
    - Recency: How recently has a customer made a purchase?
    - Frequency: How often does a customer make a purchase?
    - Monetary Value: How much money does a customer spend on a purchase?
    - A score is assigned to each customer for R, F, and M, from 1 to 5 (low to high).
    - RFM helps an organization understand the customers characteristics based on their historical transaction.
    - Based on the combination of RFM scores, customers can be grouped into any of the 11 customer segments defined by the UK Data and Marketing Association.
    - For each of the categories, marketing strategies have also been defined.
- A/B Testing:
    - A/B Testing, also known as split testing, is a method of comparing 2 versions of a webpage or app or any product against each other to determine which one performs better. It is a controlled experiment where 2 variants (A and B) are compared by testing a subject's response to A against the same subject's response to B, and determining which of the 2 variants is more effective.
- Guesstimation:
    - Guesstimation, or Guess Estimation, is the process of making an approximate or educated guess when dealing with a certain quantity, measurement or value for which precise information is not practical to obtain.
    - It involves using, intuition, knowledge, and experience to arrive at a reasonable estimate.
